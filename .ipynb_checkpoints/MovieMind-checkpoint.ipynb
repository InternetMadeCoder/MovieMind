# MovieMind — Reinforcement Learning-Based Adaptive Movie Recommendation System

## 1. Introduction and Objectives

Modern movie streaming platforms face a fundamental sequential decision-making challenge: **how to recommend movies that maximize long-term user satisfaction while preventing filter bubbles through genre diversity**. Traditional static recommenders (such as Popularity-based ranking or greedy Content-Based Filtering) treat each recommendation independently, often saturating the user's feed with repetitive items from a single dominant genre.

**MovieMind** formulates personalized movie recommendation as a **Markov Decision Process (MDP)** and solves it using Reinforcement Learning (RL).

### Key Objectives
1. **Zero-Leakage Data Pipeline**: Load and preprocess the official **MovieLens 100K** dataset from GroupLens, construct user genre preference profiles, and perform a strict **80:20 per-user train/test split** where held-out test interactions are isolated exclusively for final evaluation.
2. **Custom Gymnasium RL Environment (`MovieRecEnv`)**: Model sequential recommendation with an explicit **Markov state representation** (user genre preferences, historical interaction statistics, collaborative embeddings, and session genre exposure), action masking for already-seen movies, and a configurable **diversity-aware reward function**:
   $$R_t(s_t, a_t) = R_{\text{pref}}(u, a_t) + \lambda \cdot R_{\text{div}}(\mathbf{e}_t, a_t)$$
3. **RL Algorithms Comparison**: Implement and compare **Tabular Q-Learning** (discretized state space, explicit Q-table, Bellman updates) and **Deep Q-Network (DQN)** (PyTorch Dueling Feature-Interactive network, Experience Replay Buffer, Target Network, Huber loss, and $\varepsilon$-greedy exploration).
4. **Traditional Baselines & Rigorous Offline Evaluation**: Benchmark RL policies against **Random**, **Popularity-Based**, and **Content-Based Filtering** recommenders on held-out test data using **Precision@5, Precision@10, Recall@5, Recall@10, NDCG@5, NDCG@10, Intra-List Diversity (ILD), Genre Diversity, Catalog Coverage, and Average Cumulative Reward**.
5. **Comprehensive Ablation & Sensitivity Studies**: Analyze the impact of diversity weight $\lambda \in \{0.0, 0.10, 0.25, 0.50, 0.80\}$, exploration-exploitation schedules ($\varepsilon$ decay), and adaptive user-level recommendation trajectories.

## 3. Dataset Download and Loading

We use the official **MovieLens 100K** dataset from GroupLens Research (`https://grouplens.org/datasets/movielens/100k/`). It consists of:
- `u.data`: 100,000 explicit star ratings (1 to 5) from **943 users** on **1,682 movies** (every user has rated at least 20 movies).
- `u.item`: Movie metadata including title, release date, IMDb URL, and binary indicators across genres.
- `u.user`: User demographic information (`age`, `gender`, `occupation`, `zip_code`).

```python
MOVIELENS_100K_URL = "https://files.grouplens.org/datasets/movielens/ml-100k.zip"

ALL_GENRE_COLUMNS = [
    "unknown", "Action", "Adventure", "Animation", "Children's", "Comedy",
    "Crime", "Documentary", "Drama", "Fantasy", "Film-Noir", "Horror",
    "Musical", "Mystery", "Romance", "Sci-Fi", "Thriller", "War", "Western",
]
GENRE_NAMES = ALL_GENRE_COLUMNS[1:]  # The 18 named genres


def download_movielens_100k(data_dir: str = "data") -> str:
    """Download and extract the official MovieLens 100K dataset from GroupLens if not already present."""
    os.makedirs(data_dir, exist_ok=True)
    ml_dir = os.path.join(data_dir, "ml-100k")
    if os.path.exists(os.path.join(ml_dir, "u.data")):
        return ml_dir
    zip_path = os.path.join(data_dir, "ml-100k.zip")
    if not os.path.exists(zip_path):
        print(f"Downloading MovieLens 100K from {MOVIELENS_100K_URL}...")
        urllib.request.urlretrieve(MOVIELENS_100K_URL, zip_path)
    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(data_dir)
    return ml_dir


def load_raw_movielens(ml_dir: str) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Load ratings (u.data), movie metadata/genres (u.item), and user demographics (u.user)."""
    ratings_df = pd.read_csv(
        os.path.join(ml_dir, "u.data"),
        sep="\t",
        names=["user_id", "item_id", "rating", "timestamp"],
        engine="python",
    )
    item_cols = ["item_id", "title", "release_date", "video_release_date", "imdb_url"] + ALL_GENRE_COLUMNS
    movies_df = pd.read_csv(
        os.path.join(ml_dir, "u.item"),
        sep="|",
        names=item_cols,
        encoding="latin-1",
        engine="python",
    )
    users_df = pd.read_csv(
        os.path.join(ml_dir, "u.user"),
        sep="|",
        names=["user_id", "age", "gender", "occupation", "zip_code"],
        encoding="latin-1",
        engine="python",
    )
    return ratings_df, movies_df, users_df


ml_dir = download_movielens_100k("data")
raw_ratings_df, raw_movies_df, raw_users_df = load_raw_movielens(ml_dir)

print(f"Loaded Ratings Shape: {raw_ratings_df.shape}")
print(f"Loaded Movies Shape : {raw_movies_df.shape}")
print(f"Loaded Users Shape  : {raw_users_df.shape}")
display(raw_ratings_df.head(5))
display(raw_movies_df[["item_id", "title", "release_date"] + GENRE_NAMES[:6]].head(5))
```

## 5. Data Preprocessing and Strict Anti-Leakage 80:20 Split

### Preventing Data Leakage
To ensure a rigorous evaluation:
1. **Per-User 80:20 Split**: For every user $u$, 80% of their interactions are placed in `train_df` (80,000 ratings) and 20% are held out in `test_df` (20,000 ratings).
2. **Strict Disjointness**: We assert $\mathcal{I}_{\text{train}}(u) \cap \mathcal{I}_{\text{test}}(u) = \emptyset$ for all users.
3. **Training-Only Feature & Profile Construction**:
   - **User Genre Preference Profiles ($\mathbf{p}_u \in [0, 1]^{18}$)**: Computed *exclusively* from `train_df` ratings weighted by rating satisfaction.
   - **Item Quality & Popularity Statistics**: Bayesian-smoothed average rating and log-popularity scores computed *strictly* from `train_df`.
   - **Collaborative SVD Factors & Simulated Preference Matrix**: Low-rank SVD ($k=16$) of the training interaction matrix combined with training genre cosine similarity and Bayesian quality forms the environment's underlying user satisfaction model. Held-out `test_df` ratings are **never** accessed during training or reward computation.

```python
"""
Data Loading, Cleaning, and Preprocessing for MovieMind (MovieLens 100K).

Strict anti-leakage guarantee:
- Performs an 80:20 per-user split into train_df and test_df.
- All user preference profiles, item quality statistics, collaborative SVD factors,
  and candidate pools are computed EXCLUSIVELY from train_df.
- test_df is held out strictly for final offline recommendation evaluation.
"""

import os
import urllib.request
import zipfile
from dataclasses import dataclass
from typing import Dict, List, Set, Tuple

import numpy as np
import pandas as pd


MOVIELENS_100K_URL = "https://files.grouplens.org/datasets/movielens/ml-100k.zip"

ALL_GENRE_COLUMNS = [
    "unknown",
    "Action",
    "Adventure",
    "Animation",
    "Children's",
    "Comedy",
    "Crime",
    "Documentary",
    "Drama",
    "Fantasy",
    "Film-Noir",
    "Horror",
    "Musical",
    "Mystery",
    "Romance",
    "Sci-Fi",
    "Thriller",
    "War",
    "Western",
]

# The 18 meaningful named genres in MovieLens 100K
GENRE_NAMES = ALL_GENRE_COLUMNS[1:]


@dataclass
class PreprocessedData:
    """Container holding all cleaned datasets, encodings, and training-only features."""

    ratings_df: pd.DataFrame
    movies_df: pd.DataFrame
    users_df: pd.DataFrame
    train_df: pd.DataFrame
    test_df: pd.DataFrame
    genre_names: List[str]
    num_users: int
    num_movies: int
    num_genres: int
    # Movie matrices (num_movies, num_genres)
    movie_genre_matrix: np.ndarray
    movie_genre_norm: np.ndarray
    movie_primary_genre: np.ndarray
    # Training-derived item statistics (num_movies,)
    item_train_count: np.ndarray
    item_train_mean_rating: np.ndarray
    item_bayesian_rating: np.ndarray
    item_popularity_score: np.ndarray
    # Training-derived user profiles (num_users, num_genres) and stats (num_users, 3)
    user_genre_prefs: np.ndarray
    user_train_stats: np.ndarray
    # Training-derived user-movie preference score matrix in [0, 1] (num_users, num_movies)
    train_preference_matrix: np.ndarray
    # Training-derived latent embeddings for state/action features
    user_svd_factors: np.ndarray
    item_svd_factors: np.ndarray
    # Lookup sets per user
    train_seen_movies: Dict[int, Set[int]]
    test_relevant_movies: Dict[int, Set[int]]
    test_all_movies: Dict[int, Dict[int, float]]
    # Top candidate pool indices (computed strictly from training popularity/quality)
    candidate_pool_indices: np.ndarray


def download_movielens_100k(data_dir: str = "data") -> str:
    """Download and extract the official MovieLens 100K dataset from GroupLens if needed."""
    os.makedirs(data_dir, exist_ok=True)
    ml_dir = os.path.join(data_dir, "ml-100k")
    u_data_path = os.path.join(ml_dir, "u.data")

    if os.path.exists(u_data_path):
        return ml_dir

    zip_path = os.path.join(data_dir, "ml-100k.zip")
    if not os.path.exists(zip_path):
        print(f"Downloading MovieLens 100K from {MOVIELENS_100K_URL}...")
        urllib.request.urlretrieve(MOVIELENS_100K_URL, zip_path)

    print("Extracting MovieLens 100K dataset...")
    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(data_dir)

    return ml_dir


def load_raw_movielens(ml_dir: str) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Load ratings (u.data), movie metadata/genres (u.item), and user demographics (u.user)."""
    # 1. Load ratings
    ratings_path = os.path.join(ml_dir, "u.data")
    ratings_df = pd.read_csv(
        ratings_path,
        sep="\t",
        names=["user_id", "item_id", "rating", "timestamp"],
        engine="python",
    )

    # 2. Load movies and genres
    item_cols = ["item_id", "title", "release_date", "video_release_date", "imdb_url"] + ALL_GENRE_COLUMNS
    items_path = os.path.join(ml_dir, "u.item")
    movies_df = pd.read_csv(
        items_path,
        sep="|",
        names=item_cols,
        encoding="latin-1",
        engine="python",
    )

    # 3. Load users
    users_path = os.path.join(ml_dir, "u.user")
    users_df = pd.read_csv(
        users_path,
        sep="|",
        names=["user_id", "age", "gender", "occupation", "zip_code"],
        encoding="latin-1",
        engine="python",
    )

    return ratings_df, movies_df, users_df


def clean_and_encode_data(
    ratings_df: pd.DataFrame,
    movies_df: pd.DataFrame,
    users_df: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Clean metadata, extract release years, and encode contiguous 0-based user_idx and movie_idx."""
    movies_df = movies_df.copy()
    ratings_df = ratings_df.copy()
    users_df = users_df.copy()

    # Sort by original IDs for deterministic 0-based encoding
    movies_df = movies_df.sort_values("item_id").reset_index(drop=True)
    users_df = users_df.sort_values("user_id").reset_index(drop=True)

    movies_df["movie_idx"] = np.arange(len(movies_df), dtype=np.int64)
    users_df["user_idx"] = np.arange(len(users_df), dtype=np.int64)

    item_to_idx = dict(zip(movies_df["item_id"], movies_df["movie_idx"]))
    user_to_idx = dict(zip(users_df["user_id"], users_df["user_idx"]))

    ratings_df["user_idx"] = ratings_df["user_id"].map(user_to_idx).astype(np.int64)
    ratings_df["movie_idx"] = ratings_df["item_id"].map(item_to_idx).astype(np.int64)

    # Clean titles and extract release year
    movies_df["title"] = movies_df["title"].fillna("Unknown Title").str.strip()
    extracted_year = movies_df["title"].str.extract(r"\((\d{4})\)\s*$")[0]
    parsed_date_year = pd.to_datetime(movies_df["release_date"], errors="coerce").dt.year
    movies_df["year"] = (
        extracted_year.fillna(parsed_date_year).fillna(1995).astype(int)
    )

    # If a movie has 0 across all 18 named genres (e.g. 'unknown' only), assign Drama as fallback
    genre_mat = movies_df[GENRE_NAMES].values.astype(np.float32)
    zero_genre_mask = genre_mat.sum(axis=1) == 0
    if np.any(zero_genre_mask):
        drama_col = GENRE_NAMES.index("Drama")
        movies_df.loc[zero_genre_mask, "Drama"] = 1
        genre_mat = movies_df[GENRE_NAMES].values.astype(np.float32)

    # Build human-readable genre list and string per movie
    genre_lists = []
    genre_strs = []
    primary_genres = []
    for row in genre_mat:
        active = [GENRE_NAMES[i] for i, val in enumerate(row) if val > 0]
        genre_lists.append(active)
        genre_strs.append(" | ".join(active))
        primary_genres.append(int(np.argmax(row)))

    movies_df["genres_list"] = genre_lists
    movies_df["genres_str"] = genre_strs
    movies_df["primary_genre_idx"] = primary_genres
    movies_df["primary_genre"] = [GENRE_NAMES[idx] for idx in primary_genres]

    return ratings_df, movies_df, users_df


def split_train_test_per_user(
    ratings_df: pd.DataFrame,
    train_ratio: float = 0.80,
    seed: int = 42,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Perform an 80:20 per-user train/test split.
    Shuffles each user's ratings reproducibly with `seed` so every user has 80% in train and 20% in test.
    """
    rng = np.random.default_rng(seed)
    train_parts = []
    test_parts = []

    for _, group in ratings_df.groupby("user_idx", sort=True):
        n_interactions = len(group)
        shuffled_indices = rng.permutation(group.index.values)
        n_train = max(1, int(round(n_interactions * train_ratio)))
        if n_train >= n_interactions:
            n_train = n_interactions - 1

        train_idx = shuffled_indices[:n_train]
        test_idx = shuffled_indices[n_train:]
        train_parts.append(ratings_df.loc[train_idx])
        test_parts.append(ratings_df.loc[test_idx])

    train_df = pd.concat(train_parts, ignore_index=True)
    test_df = pd.concat(test_parts, ignore_index=True)
    return train_df, test_df


def build_preprocessed_dataset(
    data_dir: str = "data",
    train_ratio: float = 0.80,
    candidate_pool_size: int = 300,
    svd_rank: int = 16,
    seed: int = 42,
) -> PreprocessedData:
    """
    Full pipeline: download, load, clean, encode, split 80:20 per user, and construct
    user preference profiles & environment simulation parameters strictly from train_df.
    """
    ml_dir = download_movielens_100k(data_dir=data_dir)
    raw_ratings, raw_movies, raw_users = load_raw_movielens(ml_dir)
    ratings_df, movies_df, users_df = clean_and_encode_data(raw_ratings, raw_movies, raw_users)
    train_df, test_df = split_train_test_per_user(ratings_df, train_ratio=train_ratio, seed=seed)

    num_users = len(users_df)
    num_movies = len(movies_df)
    num_genres = len(GENRE_NAMES)

    # 1. Movie genre matrices
    movie_genre_matrix = movies_df[GENRE_NAMES].values.astype(np.float32)
    genre_row_norms = np.linalg.norm(movie_genre_matrix, axis=1, keepdims=True)
    genre_row_norms = np.where(genre_row_norms == 0, 1.0, genre_row_norms)
    movie_genre_norm = (movie_genre_matrix / genre_row_norms).astype(np.float32)
    movie_primary_genre = movies_df["primary_genre_idx"].values.astype(np.int64)

    # 2. Item statistics computed STRICTLY from train_df
    global_train_mean = float(train_df["rating"].mean())
    item_counts = np.zeros(num_movies, dtype=np.float32)
    item_rating_sums = np.zeros(num_movies, dtype=np.float32)

    grouped_items = train_df.groupby("movie_idx")["rating"].agg(["count", "sum"])
    item_counts[grouped_items.index.values] = grouped_items["count"].values.astype(np.float32)
    item_rating_sums[grouped_items.index.values] = grouped_items["sum"].values.astype(np.float32)

    item_train_mean = np.where(
        item_counts > 0,
        item_rating_sums / np.maximum(item_counts, 1.0),
        global_train_mean,
    ).astype(np.float32)

    # Bayesian damped rating (prior weight m = 10 ratings)
    prior_m = 10.0
    item_bayesian_rating = (
        (item_rating_sums + prior_m * global_train_mean) / (item_counts + prior_m)
    ).astype(np.float32)

    # Normalized popularity score in [0, 1] using log(1 + count)
    log_counts = np.log1p(item_counts)
    max_log_count = float(np.max(log_counts)) if np.max(log_counts) > 0 else 1.0
    item_popularity_score = (log_counts / max_log_count).astype(np.float32)

    # Attach training stats to movies_df for transparent inspection & UI
    movies_df["train_rating_count"] = item_counts.astype(int)
    movies_df["train_mean_rating"] = np.round(item_train_mean, 2)
    movies_df["bayesian_rating"] = np.round(item_bayesian_rating, 2)
    movies_df["popularity_score"] = np.round(item_popularity_score, 4)

    # 3. User preference profiles & historical stats computed STRICTLY from train_df
    user_genre_prefs = np.zeros((num_users, num_genres), dtype=np.float32)
    user_train_stats = np.zeros((num_users, 3), dtype=np.float32)

    train_seen_movies: Dict[int, Set[int]] = {u: set() for u in range(num_users)}
    test_relevant_movies: Dict[int, Set[int]] = {u: set() for u in range(num_users)}
    test_all_movies: Dict[int, Dict[int, float]] = {u: {} for u in range(num_users)}

    # Populate lookup dictionaries
    for u_idx, m_idx, rating in zip(
        train_df["user_idx"].values,
        train_df["movie_idx"].values,
        train_df["rating"].values,
    ):
        train_seen_movies[int(u_idx)].add(int(m_idx))

    for u_idx, m_idx, rating in zip(
        test_df["user_idx"].values,
        test_df["movie_idx"].values,
        test_df["rating"].values,
    ):
        u_i, m_i, r_f = int(u_idx), int(m_idx), float(rating)
        test_all_movies[u_i][m_i] = r_f
        if r_f >= 4.0:
            test_relevant_movies[u_i].add(m_i)

    # Verify strict disjointness between train and test per user
    for u in range(num_users):
        overlap = train_seen_movies[u].intersection(test_all_movies[u].keys())
        assert len(overlap) == 0, f"Data leakage detected for user {u}: {overlap}"

    # Compute user genre preference profiles from training interactions
    max_user_log_count = np.log1p(float(train_df.groupby("user_idx").size().max()))
    for u_idx, group in train_df.groupby("user_idx"):
        u = int(u_idx)
        m_indices = group["movie_idx"].values
        ratings = group["rating"].values.astype(np.float32)

        # Positive preference weight: higher ratings receive stronger positive weight
        # Map ratings {1, 2, 3, 4, 5} -> {0.1, 0.3, 0.6, 1.0, 1.4}
        weights = np.clip((ratings - 1.0) / 4.0 + 0.15, 0.1, 1.5)
        genres_watched = movie_genre_matrix[m_indices]  # (n_watched, num_genres)

        weighted_genre_sum = (genres_watched * weights[:, None]).sum(axis=0)
        genre_exposure_count = genres_watched.sum(axis=0) + 1.5  # smoothed denominator
        # Combine total affinity volume and mean rating per genre
        volume_pref = weighted_genre_sum / (np.max(weighted_genre_sum) + 1e-6)
        rate_pref = weighted_genre_sum / genre_exposure_count
        rate_pref = rate_pref / (np.max(rate_pref) + 1e-6)

        combined_pref = 0.65 * volume_pref + 0.35 * rate_pref
        if np.max(combined_pref) > 0:
            combined_pref = combined_pref / np.max(combined_pref)
        else:
            combined_pref = np.ones(num_genres, dtype=np.float32) / num_genres

        user_genre_prefs[u] = combined_pref.astype(np.float32)

        # User historical summary features: [norm_mean_rating, norm_rating_std, norm_log_activity]
        u_mean = (float(np.mean(ratings)) - 1.0) / 4.0
        u_std = float(np.std(ratings)) / 2.0
        u_act = float(np.log1p(len(ratings))) / max(max_user_log_count, 1.0)
        user_train_stats[u] = np.array([u_mean, u_std, u_act], dtype=np.float32)

    # 4. Training-only SVD Collaborative Filtering Factors (for realistic simulated user reward)
    # Construct centered/weighted implicit+explicit training matrix R_train
    interaction_matrix = np.zeros((num_users, num_movies), dtype=np.float32)
    for u_idx, m_idx, rating in zip(
        train_df["user_idx"].values,
        train_df["movie_idx"].values,
        train_df["rating"].values,
    ):
        # Implicit engagement + explicit rating satisfaction signal derived ONLY from train_df
        interaction_matrix[int(u_idx), int(m_idx)] = 0.35 + 0.65 * ((float(rating) - 1.0) / 4.0)

    # Low-rank SVD on training interaction matrix
    u_mat, s_vals, vt_mat = np.linalg.svd(interaction_matrix, full_matrices=False)
    k = min(svd_rank, len(s_vals))
    sqrt_s = np.sqrt(s_vals[:k]).astype(np.float32)
    user_svd_factors = (u_mat[:, :k] * sqrt_s[None, :]).astype(np.float32)
    item_svd_factors = (vt_mat[:k, :].T * sqrt_s[None, :]).astype(np.float32)

    svd_recon = user_svd_factors @ item_svd_factors.T  # (num_users, num_movies)
    # Normalize svd_recon per user into [0, 1]
    svd_min = np.percentile(svd_recon, 5, axis=1, keepdims=True)
    svd_max = np.percentile(svd_recon, 99, axis=1, keepdims=True)
    svd_norm = np.clip((svd_recon - svd_min) / (svd_max - svd_min + 1e-6), 0.0, 1.0)

    # Content cosine affinity between user_genre_prefs and movie_genre_norm
    user_pref_norms = np.linalg.norm(user_genre_prefs, axis=1, keepdims=True)
    user_pref_norms = np.where(user_pref_norms == 0, 1.0, user_pref_norms)
    user_genre_unit = user_genre_prefs / user_pref_norms
    content_sim = np.clip(user_genre_unit @ movie_genre_norm.T, 0.0, 1.0)

    # Item Bayesian quality normalized to [0, 1]
    bayesian_norm = np.clip((item_bayesian_rating - 1.0) / 4.0, 0.0, 1.0)

    # Combine training-derived signals into a calibrated preference score in [0, 1]
    # 45% training collaborative SVD + 35% training genre affinity + 20% training popularity/quality
    train_preference_matrix = (
        0.45 * svd_norm
        + 0.35 * content_sim
        + 0.12 * item_popularity_score[None, :]
        + 0.08 * bayesian_norm[None, :]
    ).astype(np.float32)
    train_preference_matrix = np.clip(train_preference_matrix, 0.0, 1.0)

    # If a movie was actually rated in train_df, record its true normalized rating in train_preference_matrix
    for u_idx, m_idx, rating in zip(
        train_df["user_idx"].values,
        train_df["movie_idx"].values,
        train_df["rating"].values,
    ):
        train_preference_matrix[int(u_idx), int(m_idx)] = float((rating - 1.0) / 4.0)

    # 5. Candidate pool selection (computed strictly from training popularity & quality)
    candidate_score = 0.70 * item_popularity_score + 0.30 * bayesian_norm
    pool_size = min(candidate_pool_size, num_movies)
    candidate_pool_indices = np.argsort(-candidate_score)[:pool_size].astype(np.int64)

    return PreprocessedData(
        ratings_df=ratings_df,
        movies_df=movies_df,
        users_df=users_df,
        train_df=train_df,
        test_df=test_df,
        genre_names=list(GENRE_NAMES),
        num_users=num_users,
        num_movies=num_movies,
        num_genres=num_genres,
        movie_genre_matrix=movie_genre_matrix,
        movie_genre_norm=movie_genre_norm,
        movie_primary_genre=movie_primary_genre,
        item_train_count=item_counts,
        item_train_mean_rating=item_train_mean,
        item_bayesian_rating=item_bayesian_rating,
        item_popularity_score=item_popularity_score,
        user_genre_prefs=user_genre_prefs,
        user_train_stats=user_train_stats,
        train_preference_matrix=train_preference_matrix,
        user_svd_factors=user_svd_factors,
        item_svd_factors=item_svd_factors,
        train_seen_movies=train_seen_movies,
        test_relevant_movies=test_relevant_movies,
        test_all_movies=test_all_movies,
        candidate_pool_indices=candidate_pool_indices,
    )


data = build_preprocessed_dataset("data", train_ratio=0.80, candidate_pool_size=300, seed=SEED)
print(f"Train interactions: {len(data.train_df):,} ({len(data.train_df)/len(data.ratings_df)*100:.1f}%)")
print(f"Test interactions : {len(data.test_df):,} ({len(data.test_df)/len(data.ratings_df)*100:.1f}%)")
print(f"User Genre Prefs  : {data.user_genre_prefs.shape}")
print(f"Candidate Pool    : {len(data.candidate_pool_indices)} eligible top movies")
```

```python
"""
Visualization Suite for MovieMind.
Generates publication-quality Matplotlib and Seaborn plots for Exploratory Data Analysis (EDA)
and all 7 Reinforcement Learning experimental studies.
"""

import os
from typing import Dict, List, Optional
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns



# Set consistent modern plot styling
plt.rcParams["figure.dpi"] = 120
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
plt.rcParams["font.size"] = 10


PALETTE = {
    "Random": "#8c969e",
    "Popularity": "#f59e0b",
    "Content-Based": "#10b981",
    "Tabular Q-Learning": "#3b82f6",
    "DQN (No Diversity, λ=0.0)": "#ec4899",
    "DQN (Diversity-Aware, λ=0.25)": "#8b5cf6",
    "DQN (Diversity-Aware)": "#8b5cf6",
}


def rolling_mean(values: List[float], window: int = 30) -> np.ndarray:
    """Compute smooth rolling mean with edge handling."""
    s = pd.Series(values)
    return s.rolling(window=window, min_periods=1).mean().values


def rolling_std(values: List[float], window: int = 30) -> np.ndarray:
    """Compute smooth rolling standard deviation."""
    s = pd.Series(values)
    return s.rolling(window=window, min_periods=1).std().fillna(0.0).values


def plot_eda_overview(
    data: PreprocessedData, save_path: Optional[str] = None
) -> plt.Figure:
    """
    Generate a 4-panel Exploratory Data Analysis figure:
    1. Rating distribution across 100,000 interactions
    2. Movie catalog distribution across the 18 genres
    3. User rating activity distribution (interactions per user)
    4. Average rating vs. log-popularity scatter across movies
    """
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle(
        "MovieLens 100K — Exploratory Data Analysis (EDA)",
        fontsize=15,
        fontweight="bold",
        y=0.98,
    )

    # Panel 1: Rating Distribution
    ax1 = axes[0, 0]
    rating_counts = data.ratings_df["rating"].value_counts().sort_index()
    bars = ax1.bar(
        rating_counts.index.astype(str),
        rating_counts.values,
        color=["#ef4444", "#f97316", "#eab308", "#3b82f6", "#10b981"],
        edgecolor="black",
        alpha=0.88,
    )
    ax1.set_title("1. Distribution of Explicit Movie Ratings (1–5 Stars)", fontweight="bold")
    ax1.set_xlabel("Star Rating")
    ax1.set_ylabel("Number of Ratings")
    ax1.grid(axis="y", linestyle="--", alpha=0.35)
    for bar in bars:
        h = bar.get_height()
        ax1.text(
            bar.get_x() + bar.get_width() / 2.0,
            h + 600,
            f"{int(h):,}",
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold",
        )

    # Panel 2: Genre Frequency in Movie Catalog
    ax2 = axes[0, 1]
    genre_totals = data.movie_genre_matrix.sum(axis=0)
    order = np.argsort(genre_totals)
    sorted_genres = [data.genre_names[i] for i in order]
    sorted_counts = genre_totals[order]
    colors = plt.cm.viridis(np.linspace(0.2, 0.88, len(sorted_genres)))
    ax2.barh(sorted_genres, sorted_counts, color=colors, edgecolor="black", alpha=0.88)
    ax2.set_title("2. Number of Movies per Genre in Catalog", fontweight="bold")
    ax2.set_xlabel("Number of Movies")
    ax2.grid(axis="x", linestyle="--", alpha=0.35)

    # Panel 3: User Activity Distribution
    ax3 = axes[1, 0]
    user_counts = data.ratings_df.groupby("user_idx").size()
    ax3.hist(user_counts, bins=35, color="#6366f1", edgecolor="black", alpha=0.82)
    ax3.axvline(
        user_counts.mean(),
        color="#ef4444",
        linestyle="--",
        linewidth=2,
        label=f"Mean = {user_counts.mean():.1f} ratings/user",
    )
    ax3.axvline(
        user_counts.median(),
        color="#10b981",
        linestyle=":",
        linewidth=2,
        label=f"Median = {user_counts.median():.1f} ratings/user",
    )
    ax3.set_title("3. User Interaction Activity Distribution", fontweight="bold")
    ax3.set_xlabel("Ratings per User")
    ax3.set_ylabel("Number of Users")
    ax3.legend()
    ax3.grid(axis="y", linestyle="--", alpha=0.35)

    # Panel 4: Movie Popularity vs Bayesian Rating (Training Set)
    ax4 = axes[1, 1]
    sc = ax4.scatter(
        data.item_train_count,
        data.item_bayesian_rating,
        c=data.item_popularity_score,
        cmap="plasma",
        alpha=0.65,
        edgecolors="none",
        s=28,
    )
    plt.colorbar(sc, ax=ax4, label="Normalized Popularity Score")
    ax4.set_title("4. Movie Training Rating Count vs. Bayesian Rating", fontweight="bold")
    ax4.set_xlabel("Training Rating Count")
    ax4.set_ylabel("Bayesian Smoothed Rating (Stars)")
    ax4.grid(True, linestyle="--", alpha=0.35)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


def plot_training_convergence(
    q_history: Dict[str, List[float]],
    dqn_history: Dict[str, List[float]],
    window: int = 35,
    save_path: Optional[str] = None,
) -> plt.Figure:
    """
    Plot RL training convergence:
    1. Episode Total Reward Curves (Tabular Q-Learning vs. DQN)
    2. Preference Reward vs. Diversity Bonus Evolution (DQN)
    3. Epsilon Decay Schedule (Exploration vs. Exploitation)
    4. Training Loss (DQN Huber Loss) & Tabular TD Error Curve
    """
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle(
        "Reinforcement Learning Training Dynamics — Tabular Q-Learning vs. Deep Q-Network (DQN)",
        fontsize=14,
        fontweight="bold",
        y=0.98,
    )

    episodes_q = np.arange(1, len(q_history["episode_rewards"]) + 1)
    episodes_dqn = np.arange(1, len(dqn_history["episode_rewards"]) + 1)

    # 1. Episode Reward Convergence
    ax1 = axes[0, 0]
    q_rm = rolling_mean(q_history["episode_rewards"], window)
    q_rs = rolling_std(q_history["episode_rewards"], window)
    dqn_rm = rolling_mean(dqn_history["episode_rewards"], window)
    dqn_rs = rolling_std(dqn_history["episode_rewards"], window)

    ax1.plot(episodes_q, q_rm, color="#3b82f6", linewidth=2.2, label="Tabular Q-Learning (Rolling Mean)")
    ax1.fill_between(episodes_q, q_rm - 0.5 * q_rs, q_rm + 0.5 * q_rs, color="#3b82f6", alpha=0.18)
    ax1.plot(episodes_dqn, dqn_rm, color="#8b5cf6", linewidth=2.4, label="Deep Q-Network (DQN) (Rolling Mean)")
    ax1.fill_between(episodes_dqn, dqn_rm - 0.5 * dqn_rs, dqn_rm + 0.5 * dqn_rs, color="#8b5cf6", alpha=0.20)
    ax1.set_title("1. Episode Cumulative Reward Convergence", fontweight="bold")
    ax1.set_xlabel("Training Episode")
    ax1.set_ylabel("Cumulative Episode Reward")
    ax1.legend(loc="lower right")
    ax1.grid(True, linestyle="--", alpha=0.35)

    # 2. Preference Reward & Diversity Bonus Decomposition
    ax2 = axes[0, 1]
    ax2.plot(
        episodes_dqn,
        rolling_mean(dqn_history["preference_rewards"], window),
        color="#10b981",
        linewidth=2.2,
        label="DQN Preference Reward",
    )
    ax2.plot(
        episodes_q,
        rolling_mean(q_history["preference_rewards"], window),
        color="#10b981",
        linestyle="--",
        linewidth=1.8,
        label="Q-Learning Preference Reward",
    )
    ax2.plot(
        episodes_dqn,
        rolling_mean(dqn_history["diversity_bonuses"], window),
        color="#f59e0b",
        linewidth=2.2,
        label="DQN Diversity Bonus (Unweighted)",
    )
    ax2.plot(
        episodes_q,
        rolling_mean(q_history["diversity_bonuses"], window),
        color="#f59e0b",
        linestyle="--",
        linewidth=1.8,
        label="Q-Learning Diversity Bonus (Unweighted)",
    )
    ax2.set_title("2. Preference Satisfaction vs. Genre Diversity Bonus", fontweight="bold")
    ax2.set_xlabel("Training Episode")
    ax2.set_ylabel("Cumulative Component Score")
    ax2.legend(loc="lower right", fontsize=9)
    ax2.grid(True, linestyle="--", alpha=0.35)

    # 3. Epsilon Decay Schedule
    ax3 = axes[1, 0]
    ax3.plot(episodes_dqn, dqn_history["epsilons"], color="#ef4444", linewidth=2.4, label="Epsilon (ε) Schedule")
    ax3_twin = ax3.twinx()
    ax3_twin.plot(
        episodes_dqn,
        dqn_rm,
        color="#8b5cf6",
        linestyle=":",
        linewidth=2.0,
        label="DQN Reward",
    )
    ax3.set_title("3. Exploration-Exploitation Trade-off (Epsilon Decay)", fontweight="bold")
    ax3.set_xlabel("Training Episode")
    ax3.set_ylabel("Exploration Rate (ε)", color="#ef4444")
    ax3_twin.set_ylabel("DQN Rolling Reward", color="#8b5cf6")
    ax3.grid(True, linestyle="--", alpha=0.35)

    # 4. Training Loss & TD Error Curves
    ax4 = axes[1, 1]
    loss_rm = rolling_mean(dqn_history["losses"], window)
    td_rm = rolling_mean(q_history["td_errors"], window)
    ax4.plot(episodes_dqn, loss_rm, color="#8b5cf6", linewidth=2.2, label="DQN Smooth L1 (Huber) Loss")
    ax4_twin = ax4.twinx()
    ax4_twin.plot(episodes_q, td_rm, color="#3b82f6", linestyle="--", linewidth=2.0, label="Tabular Mean |TD Error|")
    ax4.set_title("4. DQN Bellman Loss & Tabular Q-Learning TD Error", fontweight="bold")
    ax4.set_xlabel("Training Episode")
    ax4.set_ylabel("DQN Huber Loss", color="#8b5cf6")
    ax4_twin.set_ylabel("Tabular |TD Error|", color="#3b82f6")
    ax4.grid(True, linestyle="--", alpha=0.35)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


def plot_ranking_and_algorithm_comparison(
    eval_df: pd.DataFrame, save_path: Optional[str] = None
) -> plt.Figure:
    """
    Generate a 4-panel comparison across all models:
    1. Precision@5, Recall@5, NDCG@5 on Held-Out Test Split
    2. Precision@10, Recall@10, NDCG@10 on Held-Out Test Split
    3. Intra-List Diversity (ILD), Genre Diversity, and Catalog Coverage
    4. Average Cumulative Reward (Decomposed into Preference + λ*Diversity)
    """
    fig, axes = plt.subplots(2, 2, figsize=(16, 11))
    fig.suptitle(
        "Comprehensive Model Evaluation — Held-Out Ranking Accuracy, Diversity & Cumulative Reward",
        fontsize=14,
        fontweight="bold",
        y=0.98,
    )

    models = eval_df["Model"].tolist()
    x = np.arange(len(models))
    width = 0.25

    # Panel 1: Top-5 Held-Out Ranking Metrics
    ax1 = axes[0, 0]
    ax1.bar(x - width, eval_df["Precision@5"], width, label="Precision@5", color="#3b82f6", edgecolor="black")
    ax1.bar(x, eval_df["Recall@5"], width, label="Recall@5", color="#10b981", edgecolor="black")
    ax1.bar(x + width, eval_df["NDCG@5"], width, label="NDCG@5", color="#8b5cf6", edgecolor="black")
    ax1.set_title("1. Held-Out Top-5 Ranking Metrics (Precision@5, Recall@5, NDCG@5)", fontweight="bold")
    ax1.set_xticks(x)
    ax1.set_xticklabels(models, rotation=18, ha="right")
    ax1.set_ylabel("Metric Score")
    ax1.legend()
    ax1.grid(axis="y", linestyle="--", alpha=0.35)

    # Panel 2: Top-10 Held-Out Ranking Metrics
    ax2 = axes[0, 1]
    ax2.bar(x - width, eval_df["Precision@10"], width, label="Precision@10", color="#2563eb", edgecolor="black")
    ax2.bar(x, eval_df["Recall@10"], width, label="Recall@10", color="#059669", edgecolor="black")
    ax2.bar(x + width, eval_df["NDCG@10"], width, label="NDCG@10", color="#7c3aed", edgecolor="black")
    ax2.set_title("2. Held-Out Top-10 Ranking Metrics (Precision@10, Recall@10, NDCG@10)", fontweight="bold")
    ax2.set_xticks(x)
    ax2.set_xticklabels(models, rotation=18, ha="right")
    ax2.set_ylabel("Metric Score")
    ax2.legend()
    ax2.grid(axis="y", linestyle="--", alpha=0.35)

    # Panel 3: Diversity & Catalog Coverage Comparison
    ax3 = axes[1, 0]
    ax3.bar(
        x - width,
        eval_df["Intra-List Diversity"],
        width,
        label="Intra-List Diversity (ILD)",
        color="#f59e0b",
        edgecolor="black",
    )
    ax3.bar(
        x,
        eval_df["Genre Diversity"],
        width,
        label="Genre Coverage Ratio",
        color="#ec4899",
        edgecolor="black",
    )
    ax3.bar(
        x + width,
        eval_df["Catalog Coverage"],
        width,
        label="Catalog Coverage",
        color="#06b6d4",
        edgecolor="black",
    )
    ax3.set_title("3. Recommendation Diversity & Catalog Coverage", fontweight="bold")
    ax3.set_xticks(x)
    ax3.set_xticklabels(models, rotation=18, ha="right")
    ax3.set_ylabel("Score in [0, 1]")
    ax3.legend()
    ax3.grid(axis="y", linestyle="--", alpha=0.35)

    # Panel 4: Simulated Environment Cumulative Reward
    ax4 = axes[1, 1]
    pref_vals = eval_df["Avg Preference Reward"].values
    tot_vals = eval_df["Avg Cumulative Reward"].values
    div_contrib = np.maximum(0.0, tot_vals - pref_vals)
    ax4.bar(x, pref_vals, width=0.48, label="Preference Satisfaction Component", color="#10b981", edgecolor="black")
    ax4.bar(
        x,
        div_contrib,
        width=0.48,
        bottom=pref_vals,
        label="Weighted Diversity Bonus (λ · R_div)",
        color="#8b5cf6",
        edgecolor="black",
    )
    for i, tot in enumerate(tot_vals):
        ax4.text(i, tot + 0.12, f"{tot:.2f}", ha="center", va="bottom", fontweight="bold", fontsize=9)
    ax4.set_title("4. Average Cumulative Episode Reward Decomposition", fontweight="bold")
    ax4.set_xticks(x)
    ax4.set_xticklabels(models, rotation=18, ha="right")
    ax4.set_ylabel("Cumulative Episode Reward (K=10)")
    ax4.set_ylim(0, max(tot_vals) * 1.15)
    ax4.legend()
    ax4.grid(axis="y", linestyle="--", alpha=0.35)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


def plot_diversity_weight_ablation(
    lambda_df: pd.DataFrame, save_path: Optional[str] = None
) -> plt.Figure:
    """
    Plot the impact of varying the diversity weight lambda:
    1. Preference Reward vs. Intra-List Diversity & Genre Diversity across lambda
    2. Held-out NDCG@10 & Precision@10 vs. Catalog Coverage across lambda
    """
    fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))
    fig.suptitle(
        "Sensitivity & Ablation Study — Impact of Diversity Weight (λ) on DQN Recommendations",
        fontsize=14,
        fontweight="bold",
        y=1.02,
    )

    lambdas = lambda_df["Lambda"].values

    # Panel 1: Trade-off between Preference Satisfaction and Diversity
    ax1 = axes[0]
    ax1.plot(
        lambdas,
        lambda_df["Avg Preference Reward"] / 10.0,
        marker="o",
        linewidth=2.4,
        color="#10b981",
        label="Per-Step Preference Reward",
    )
    ax1.plot(
        lambdas,
        lambda_df["Intra-List Diversity"],
        marker="s",
        linewidth=2.4,
        color="#f59e0b",
        label="Intra-List Diversity (ILD)",
    )
    ax1.plot(
        lambdas,
        lambda_df["Genre Diversity"],
        marker="^",
        linewidth=2.4,
        color="#8b5cf6",
        label="Genre Coverage Ratio",
    )
    ax1.axvline(0.25, color="#ef4444", linestyle="--", alpha=0.7, label="Balanced Operating Point (λ=0.25)")
    ax1.set_title("1. Preference Alignment vs. Intra-List & Genre Diversity", fontweight="bold")
    ax1.set_xlabel("Diversity Weight (λ)")
    ax1.set_ylabel("Normalized Score in [0, 1]")
    ax1.legend()
    ax1.grid(True, linestyle="--", alpha=0.35)

    # Panel 2: Held-out Ranking Quality & Catalog Coverage across Lambda
    ax2 = axes[1]
    ax2.plot(
        lambdas,
        lambda_df["NDCG@10"],
        marker="o",
        linewidth=2.4,
        color="#3b82f6",
        label="Held-Out NDCG@10",
    )
    ax2.plot(
        lambdas,
        lambda_df["Precision@10"],
        marker="d",
        linewidth=2.4,
        color="#ec4899",
        label="Held-Out Precision@10",
    )
    ax2.plot(
        lambdas,
        lambda_df["Catalog Coverage"],
        marker="s",
        linewidth=2.4,
        color="#06b6d4",
        label="Catalog Coverage",
    )
    ax2.axvline(0.25, color="#ef4444", linestyle="--", alpha=0.7, label="Balanced Operating Point (λ=0.25)")
    ax2.set_title("2. Held-Out Accuracy (NDCG@10, Precision@10) & Catalog Coverage", fontweight="bold")
    ax2.set_xlabel("Diversity Weight (λ)")
    ax2.set_ylabel("Metric Value")
    ax2.legend()
    ax2.grid(True, linestyle="--", alpha=0.35)

    plt.tight_layout()
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


def plot_genre_distribution_comparison(
    data: PreprocessedData,
    model_recs: Dict[str, Dict[int, List[int]]],
    save_path: Optional[str] = None,
) -> plt.Figure:
    """
    Compare the distribution of recommended genres across models to illustrate how
    diversity-aware RL prevents single-genre collapse and improves long-tail genre exposure.
    """
    genre_names = data.genre_names
    records = []

    for model_name, user_rec_dict in model_recs.items():
        genre_counts = np.zeros(data.num_genres, dtype=np.float64)
        for recs in user_rec_dict.values():
            genre_counts += data.movie_genre_matrix[recs].sum(axis=0)
        total = max(1.0, float(np.sum(genre_counts)))
        genre_shares = (genre_counts / total) * 100.0
        for g_idx, g_name in enumerate(genre_names):
            records.append(
                {
                    "Model": model_name,
                    "Genre": g_name,
                    "Share (%)": genre_shares[g_idx],
                }
            )

    df_genre = pd.DataFrame(records)
    fig, ax = plt.subplots(figsize=(15, 6))
    sns.barplot(
        data=df_genre,
        x="Genre",
        y="Share (%)",
        hue="Model",
        ax=ax,
        edgecolor="black",
    )
    ax.set_title(
        "Recommended Genre Exposure Distribution Across Algorithms (% of Total Genre Tags)",
        fontsize=13,
        fontweight="bold",
    )
    ax.set_xlabel("Movie Genre")
    ax.set_ylabel("Share of Recommended Genre Tags (%)")
    ax.tick_params(axis="x", rotation=30)
    ax.legend(title="Algorithm", loc="upper right")
    ax.grid(axis="y", linestyle="--", alpha=0.35)

    plt.tight_layout()
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


def plot_epsilon_exploration_study(
    decay_histories: Dict[str, Dict[str, List[float]]],
    window: int = 35,
    save_path: Optional[str] = None,
) -> plt.Figure:
    """
    Plot the Exploration vs. Exploitation study across different epsilon decay schedules.
    """
    fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))
    fig.suptitle(
        "Exploration vs. Exploitation Analysis — Impact of Epsilon (ε) Decay Schedules",
        fontsize=14,
        fontweight="bold",
        y=1.02,
    )

    colors = ["#ef4444", "#8b5cf6", "#10b981", "#f59e0b"]
    for idx, (label, hist) in enumerate(decay_histories.items()):
        c = colors[idx % len(colors)]
        eps = np.arange(1, len(hist["epsilons"]) + 1)
        axes[0].plot(eps, hist["epsilons"], label=label, color=c, linewidth=2.2)
        axes[1].plot(
            eps,
            rolling_mean(hist["episode_rewards"], window),
            label=label,
            color=c,
            linewidth=2.2,
        )

    axes[0].set_title("1. Exploration Rate (ε) Decay Trajectories", fontweight="bold")
    axes[0].set_xlabel("Training Episode")
    axes[0].set_ylabel("Epsilon (ε)")
    axes[0].legend()
    axes[0].grid(True, linestyle="--", alpha=0.35)

    axes[1].set_title("2. Resulting Episode Reward Convergence", fontweight="bold")
    axes[1].set_xlabel("Training Episode")
    axes[1].set_ylabel("Rolling Mean Episode Reward")
    axes[1].legend()
    axes[1].grid(True, linestyle="--", alpha=0.35)

    plt.tight_layout()
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fig.savefig(save_path, bbox_inches="tight")
    return fig


fig_eda = plot_eda_overview(data, save_path="figures/01_eda_overview.png")
plt.show()
```

## 6. Reinforcement Learning Environment Implementation (`MovieRecEnv`)

We implement a custom Gymnasium-compatible environment (`MovieRecEnv`) that models a sequential recommendation session of length $K=10$ for a target user $u$.

### Markov State Representation ($s_t \in [-1, 1]^{57}$)
At step $t$, the state vector $s_t$ concatenates four components:
1. **User Genre Preference Vector ($\mathbf{p}_{u,t} \in [0, 1]^{18}$)**: Dynamic user affinity across the 18 genres, initialized from training interactions and updated online as the user reacts to recommendations.
2. **Historical & Session Interaction Features ($\mathbf{h}_{u,t} \in [-1, 1]^{20}$)**: Normalized user mean rating, rating standard deviation, log interaction count, running session preference reward, and the user's 16-dimensional training SVD collaborative embedding.
3. **Session Genre Exposure Vector ($\mathbf{e}_t \in [0, 1]^{18}$)**: Fraction of recommendations made so far in the current episode belonging to each genre.
4. **Episode Step Progress ($t / K \in [0, 1]^1$)**: Normalized horizon indicator.

**Why $s_t$ satisfies the Markov Property**: Both the immediate reward $R(s_t, a_t)$ (which depends on user preference $\mathbf{p}_{u,t}$ and current genre saturation $\mathbf{e}_t$) and the next state transition $s_{t+1} = f(s_t, a_t)$ are completely determined by $(s_t, a_t)$ without requiring the raw sequence of past actions.

### Action Space & Candidate Masking
- `action_space = spaces.Discrete(num_movies)`
- At every step, `get_action_mask()` masks out all movies the user already rated in `train_df` **and** all movies already recommended earlier in the current episode.

### Diversity-Aware Reward Function
$$R_t(s_t, a_t) = R_{\text{pref}}(u, a_t) + \lambda \cdot R_{\text{div}}(\mathbf{e}_t, a_t)$$
where $R_{\text{pref}}(u, a_t) \in [0, 1]$ measures training-derived user satisfaction and $R_{\text{div}}(\mathbf{e}_t, a_t) \in [0, 1]$ rewards session genre novelty and intra-list cosine dissimilarity, softly gated by user relevance so diversity never overwhelms core user preferences.

```python
"""
Custom Gymnasium-Compatible Reinforcement Learning Environment for MovieMind.

Markov Decision Process (MDP) Formulation:
- State s_t (57-dim continuous vector):
    1. User genre preference profile p_{u,t} (18 dims in [0, 1])
    2. User historical & session interaction features h_{u,t} (20 dims):
       - [norm_mean_rating, norm_rating_std, norm_log_activity, session_avg_pref_reward] (4 dims)
       - Training-derived historical collaborative interaction embedding u_svd (16 dims)
    3. Session genre exposure vector e_t (18 dims): fraction of recommendations so far in each genre
    4. Normalized episode step progress t / T_max (1 dim)
- Action a_t:
    Discrete movie index selected from the eligible candidate pool (excluding movies
    already rated by the user in train_df and movies already recommended in the episode).
- Reward r_t:
    Total Reward = Preference Reward + lambda_div * Diversity Bonus
- Episode Termination:
    Terminates after `episode_length` (default K=10) recommendations.
"""

from typing import Any, Dict, List, Optional, Tuple

import gymnasium as gym
from gymnasium import spaces
import numpy as np



class MovieRecEnv(gym.Env):
    """
    Gymnasium-compatible adaptive movie recommendation environment.
    Supports both continuous state vectors (for DQN) and discretized state indices (for Tabular Q-Learning).
    """

    metadata = {"render_modes": ["human"]}

    def __init__(
        self,
        data: PreprocessedData,
        episode_length: int = 10,
        diversity_lambda: float = 0.25,
        use_candidate_pool: bool = True,
        user_pool: Optional[List[int]] = None,
        seed: int = 42,
    ):
        super().__init__()
        self.data = data
        self.episode_length = int(episode_length)
        self.diversity_lambda = float(diversity_lambda)
        self.use_candidate_pool = bool(use_candidate_pool)
        self.user_pool = (
            list(user_pool) if user_pool is not None else list(range(data.num_users))
        )
        self.rng = np.random.default_rng(seed)

        self.num_users = data.num_users
        self.num_movies = data.num_movies
        self.num_genres = data.num_genres
        self.svd_dim = data.user_svd_factors.shape[1]

        # Pre-normalize user SVD factors to [-1, 1] for state stability
        max_svd = float(np.max(np.abs(data.user_svd_factors))) + 1e-6
        self.norm_user_svd = (data.user_svd_factors / max_svd).astype(np.float32)

        # State vector dimension:
        # 18 (user genre prefs) + 4 (user stats & session reward) + 16 (historical SVD features)
        # + 18 (session genre exposure) + 1 (step progress) = 57
        self.state_dim = self.num_genres * 2 + 5 + self.svd_dim
        self.observation_space = spaces.Box(
            low=-1.0,
            high=1.0,
            shape=(self.state_dim,),
            dtype=np.float32,
        )
        self.action_space = spaces.Discrete(self.num_movies)

        # Discretized state space size for Tabular Q-Learning:
        # 18 (top preferred genre) * 19 (most exposed session genre or 18='none') * 2 (saturation flag) = 684 states
        self.num_discrete_states = self.num_genres * (self.num_genres + 1) * 2

        # Episode state variables
        self.current_user: int = 0
        self.current_step: int = 0
        self.current_user_prefs: np.ndarray = np.zeros(self.num_genres, dtype=np.float32)
        self.session_genre_counts: np.ndarray = np.zeros(self.num_genres, dtype=np.float32)
        self.recommended_movies: List[int] = []
        self.session_pref_rewards: List[float] = []
        self.session_div_bonuses: List[float] = []
        self._eligible_mask: np.ndarray = np.ones(self.num_movies, dtype=bool)

    def _init_eligible_mask(self, user_idx: int) -> None:
        """Initialize candidate mask excluding training-seen movies for `user_idx`."""
        if self.use_candidate_pool:
            mask = np.zeros(self.num_movies, dtype=bool)
            mask[self.data.candidate_pool_indices] = True
        else:
            mask = np.ones(self.num_movies, dtype=bool)

        # Mask out movies already seen by the user in train_df
        seen = self.data.train_seen_movies.get(user_idx, set())
        if seen:
            seen_arr = np.fromiter(seen, dtype=np.int64, count=len(seen))
            mask[seen_arr] = False

        # Fallback if a very active user has seen almost the entire top pool
        if np.sum(mask) < self.episode_length * 2:
            mask = np.ones(self.num_movies, dtype=bool)
            if seen:
                seen_arr = np.fromiter(seen, dtype=np.int64, count=len(seen))
                mask[seen_arr] = False

        self._eligible_mask = mask

    def get_action_mask(self) -> np.ndarray:
        """Return boolean mask of shape (num_movies,) indicating currently eligible movies."""
        return self._eligible_mask.copy()

    def get_eligible_actions(self) -> np.ndarray:
        """Return 1D array of currently eligible movie indices."""
        return np.flatnonzero(self._eligible_mask)

    def _get_obs(self) -> np.ndarray:
        """Construct the 57-dimensional Markov state representation."""
        user_stats = self.data.user_train_stats[self.current_user]  # (3,)
        avg_session_pref = (
            float(np.mean(self.session_pref_rewards))
            if self.session_pref_rewards
            else float(user_stats[0])
        )
        interaction_stats = np.array(
            [user_stats[0], user_stats[1], user_stats[2], avg_session_pref],
            dtype=np.float32,
        )
        user_svd = self.norm_user_svd[self.current_user]  # (16,)

        if self.current_step > 0:
            genre_exposure = np.clip(
                self.session_genre_counts / float(self.current_step), 0.0, 1.0
            )
        else:
            genre_exposure = np.zeros(self.num_genres, dtype=np.float32)

        step_progress = np.array(
            [min(1.0, float(self.current_step) / max(1.0, float(self.episode_length)))],
            dtype=np.float32,
        )

        obs = np.concatenate(
            [
                self.current_user_prefs,
                interaction_stats,
                user_svd,
                genre_exposure,
                step_progress,
            ],
            axis=0,
        ).astype(np.float32)
        return np.clip(obs, -1.0, 1.0)

    def discretize_state(self, obs: Optional[np.ndarray] = None) -> int:
        """
        Convert continuous Markov state into a discrete state index for Tabular Q-Learning.
        Captures:
          1. User's dominant preferred genre (0..17)
          2. Session's most heavily exposed genre so far (0..17, or 18 if step == 0)
          3. Saturation indicator: 1 if dominant preferred genre already exceeds 35% of session recommendations
        """
        if obs is None:
            obs = self._get_obs()

        user_prefs = obs[: self.num_genres]
        exposure_start = self.num_genres + 4 + self.svd_dim
        genre_exposure = obs[exposure_start : exposure_start + self.num_genres]

        top_user_genre = int(np.argmax(user_prefs))
        if np.max(genre_exposure) <= 1e-5:
            most_exposed_genre = self.num_genres  # 18 = no exposure yet
            is_saturated = 0
        else:
            most_exposed_genre = int(np.argmax(genre_exposure))
            is_saturated = 1 if genre_exposure[top_user_genre] >= 0.35 else 0

        state_idx = (
            top_user_genre * ((self.num_genres + 1) * 2)
            + most_exposed_genre * 2
            + is_saturated
        )
        return int(state_idx)

    def compute_reward_components(self, movie_idx: int) -> Tuple[float, float, float]:
        """
        Compute (preference_reward, diversity_bonus, total_reward) for recommending `movie_idx`
        in the current state without mutating state.
        """
        # 1. Simulated Preference Reward derived strictly from training profiles
        base_train_pref = float(self.data.train_preference_matrix[self.current_user, movie_idx])
        movie_g_norm = self.data.movie_genre_norm[movie_idx]
        user_p_norm = self.current_user_prefs / (np.linalg.norm(self.current_user_prefs) + 1e-6)
        dynamic_genre_match = float(np.clip(np.dot(user_p_norm, movie_g_norm), 0.0, 1.0))

        pref_reward = 0.75 * base_train_pref + 0.25 * dynamic_genre_match

        # 2. Genre Diversity Bonus
        movie_genres = self.data.movie_genre_matrix[movie_idx]
        active_genres = np.flatnonzero(movie_genres > 0)

        if self.current_step == 0 or len(self.recommended_movies) == 0:
            session_novelty = 1.0
            intra_list_dissim = 1.0
        else:
            genre_exposure = self.session_genre_counts / float(self.current_step)
            avg_exposure_of_movie_genres = float(np.mean(genre_exposure[active_genres]))
            session_novelty = float(np.clip(1.0 - 1.8 * avg_exposure_of_movie_genres, 0.0, 1.0))

            prev_norms = self.data.movie_genre_norm[self.recommended_movies]
            sims = prev_norms @ movie_g_norm
            intra_list_dissim = float(np.clip(1.0 - np.mean(sims), 0.0, 1.0))

        raw_diversity = 0.55 * session_novelty + 0.45 * intra_list_dissim
        # Gate diversity bonus mildly by user genre relevance so irrelevant items do not dominate
        relevance_gate = 0.35 + 0.65 * dynamic_genre_match
        diversity_bonus = float(np.clip(raw_diversity * relevance_gate, 0.0, 1.0))

        total_reward = float(pref_reward + self.diversity_lambda * diversity_bonus)
        return pref_reward, diversity_bonus, total_reward

    def reset(
        self,
        *,
        seed: Optional[int] = None,
        options: Optional[Dict[str, Any]] = None,
    ) -> Tuple[np.ndarray, Dict[str, Any]]:
        """Reset environment for a new user episode."""
        super().reset(seed=seed)
        if seed is not None:
            self.rng = np.random.default_rng(seed)

        if options is not None and "user_idx" in options:
            self.current_user = int(options["user_idx"])
        else:
            self.current_user = int(self.rng.choice(self.user_pool))

        self.current_step = 0
        self.current_user_prefs = self.data.user_genre_prefs[self.current_user].copy()
        self.session_genre_counts = np.zeros(self.num_genres, dtype=np.float32)
        self.recommended_movies = []
        self.session_pref_rewards = []
        self.session_div_bonuses = []
        self._init_eligible_mask(self.current_user)

        obs = self._get_obs()
        info = {
            "user_idx": self.current_user,
            "discrete_state": self.discretize_state(obs),
            "eligible_count": int(np.sum(self._eligible_mask)),
        }
        return obs, info

    def step(
        self, action: int
    ) -> Tuple[np.ndarray, float, bool, bool, Dict[str, Any]]:
        """Execute movie recommendation action `action` and return (next_obs, reward, terminated, truncated, info)."""
        movie_idx = int(action)

        # Enforce candidate masking if an invalid/already-seen movie was passed
        already_used_penalty = 0.0
        if not self._eligible_mask[movie_idx]:
            eligible = self.get_eligible_actions()
            if len(eligible) > 0:
                movie_idx = int(self.rng.choice(eligible))
            already_used_penalty = -0.15

        pref_reward, div_bonus, total_reward = self.compute_reward_components(movie_idx)
        total_reward += already_used_penalty

        # Update session state and adaptive user preferences based on simulated feedback
        movie_genres = self.data.movie_genre_matrix[movie_idx]
        self.session_genre_counts += movie_genres
        self.recommended_movies.append(movie_idx)
        self.session_pref_rewards.append(pref_reward)
        self.session_div_bonuses.append(div_bonus)
        self._eligible_mask[movie_idx] = False
        self.current_step += 1

        # Adaptive preference update: positive satisfaction gently reinforces movie's genres
        feedback_delta = (pref_reward - 0.55) * 0.08
        self.current_user_prefs = np.clip(
            self.current_user_prefs + feedback_delta * movie_genres,
            0.05,
            1.0,
        )
        if np.max(self.current_user_prefs) > 0:
            self.current_user_prefs = self.current_user_prefs / np.max(self.current_user_prefs)

        terminated = bool(
            self.current_step >= self.episode_length
            or np.sum(self._eligible_mask) == 0
        )
        truncated = False

        next_obs = self._get_obs()
        info = {
            "user_idx": self.current_user,
            "movie_idx": movie_idx,
            "preference_reward": pref_reward,
            "diversity_bonus": div_bonus,
            "total_reward": total_reward,
            "discrete_state": self.discretize_state(next_obs),
            "step": self.current_step,
        }
        return next_obs, float(total_reward), terminated, truncated, info


# Instantiate and verify the Gymnasium environment
env_demo = MovieRecEnv(data=data, episode_length=10, diversity_lambda=0.25, seed=SEED)
obs_sample, info_sample = env_demo.reset(seed=SEED)

print(f"Observation Space : {env_demo.observation_space}")
print(f"Action Space      : {env_demo.action_space}")
print(f"Discrete States   : {env_demo.num_discrete_states} states (for Tabular Q-Learning)")
print(f"Sample User Index : {info_sample['user_idx']} | Eligible Unseen Candidates: {info_sample['eligible_count']}")

eligible_demo = env_demo.get_eligible_actions()
next_obs_demo, r_demo, term_demo, trunc_demo, step_info_demo = env_demo.step(int(eligible_demo[0]))
print(
    f"Sample Step -> Movie ID: {step_info_demo['movie_idx']} | "
    f"Pref Reward: {step_info_demo['preference_reward']:.3f} | "
    f"Diversity Bonus: {step_info_demo['diversity_bonus']:.3f} | "
    f"Total Reward: {r_demo:.3f}"
)
```

## 7. Tabular Q-Learning Implementation

Our **Tabular Q-Learning** agent (`TabularQLearningAgent`) operates on a discretized state representation:
- **State Discretization**: Maps continuous state $s_t$ to $s_{\text{disc}} \in \{0, \dots, 683\}$ based on the tuple $(\text{dominant\_user\_genre}, \text{most\_exposed\_session\_genre}, \text{is\_dominant\_genre\_saturated})$.
- **Q-Table Structure**: Maintains an explicit state-action table $Q(s_{\text{disc}}, a)$ alongside a state-genre aggregated table $Q_{\text{genre}}(s_{\text{disc}}, g)$ to share statistical strength across movies sharing genres.
- **Bellman Update Rule**:
  $$Q(s_{\text{disc}}, a) \leftarrow Q(s_{\text{disc}}, a) + \alpha \left[ r_t + \gamma \max_{a' \in \mathcal{A}_{\text{eligible}}(s'_{t+1})} Q(s'_{\text{disc}}, a') - Q(s_{\text{disc}}, a) \right]$$

```python
"""
Reinforcement Learning Agents for MovieMind:
1. Tabular Q-Learning Agent (Discretized state representation, Q-table, Bellman updates, epsilon-greedy)
2. Deep Q-Network (DQN) Agent (PyTorch neural network, Experience Replay Buffer, Target Network, Huber loss)
"""

from typing import Dict, List, Optional, Tuple

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim



class TabularQLearningAgent:
    """
    Tabular Q-Learning agent operating on the discretized Markov state space.
    Maintains an explicit state-action Q-table Q(s_disc, a) and state-genre Q-table
    updated via Tabular Bellman equations with epsilon-greedy exploration.
    """

    def __init__(
        self,
        num_states: int,
        num_actions: int,
        data: PreprocessedData,
        lr: float = 0.20,
        gamma: float = 0.85,
        epsilon_start: float = 1.0,
        epsilon_end: float = 0.05,
        epsilon_decay: float = 0.995,
        seed: int = 42,
    ):
        self.num_states = int(num_states)
        self.num_actions = int(num_actions)
        self.num_genres = data.num_genres
        self.lr = float(lr)
        self.gamma = float(gamma)
        self.epsilon = float(epsilon_start)
        self.epsilon_start = float(epsilon_start)
        self.epsilon_end = float(epsilon_end)
        self.epsilon_decay = float(epsilon_decay)
        self.rng = np.random.default_rng(seed)

        self.movie_genre_matrix = data.movie_genre_matrix.astype(np.float32)
        self.movie_genre_norm = data.movie_genre_norm.astype(np.float32)
        genre_row_sums = np.maximum(1.0, self.movie_genre_matrix.sum(axis=1, keepdims=True))
        self.movie_genre_l1 = (self.movie_genre_matrix / genre_row_sums).astype(np.float32)
        self.movie_primary_genre = data.movie_primary_genre.astype(np.int64)

        # Training-derived item quality prior (strictly from train_df)
        bayesian_norm = np.clip((data.item_bayesian_rating - 1.0) / 4.0, 0.0, 1.0)
        self.item_prior = (
            0.70 * data.item_popularity_score + 0.30 * bayesian_norm
        ).astype(np.float32)

        # Calibrate Q-table initialization to the Bellman fixed-point structure:
        # Q*(s, a) = E[r(s, a)] + (gamma / (1 - gamma)) * r_max
        r_max_est = 0.82
        continuation_val = (self.gamma / (1.0 - self.gamma)) * r_max_est
        base_step_r = 0.62 * self.item_prior + 0.18
        self.q_table = np.tile(
            (base_step_r + continuation_val)[None, :],
            (self.num_states, 1),
        ).astype(np.float32)
        genre_base_r = (self.movie_genre_l1.T @ base_step_r) / np.maximum(
            1.0, self.movie_genre_l1.sum(axis=0)
        )
        self.genre_q_table = np.tile(
            (genre_base_r + continuation_val)[None, :],
            (self.num_states, 1),
        ).astype(np.float32)
        self.visit_counts = np.zeros((self.num_states, self.num_actions), dtype=np.float32)

        # Encode structural prior into genre_q_table based on discretized state semantics
        for s_idx in range(self.num_states):
            top_u_genre = s_idx // ((self.num_genres + 1) * 2)
            rem = s_idx % ((self.num_genres + 1) * 2)
            exposed_genre = rem // 2
            is_saturated = rem % 2
            self.genre_q_table[s_idx, top_u_genre] += 0.15 * (
                1.0 - 0.55 * is_saturated
            )
            if exposed_genre < self.num_genres:
                self.genre_q_table[s_idx, exposed_genre] -= (
                    0.12 * (0.5 + 0.5 * is_saturated)
                )

    def _tabular_q(self, discrete_state: int) -> np.ndarray:
        """Pure tabular Q-estimate from state-action and state-genre Q-tables."""
        direct_q = self.q_table[discrete_state]
        genre_q = self.movie_genre_l1 @ self.genre_q_table[discrete_state]
        return 0.72 * direct_q + 0.28 * genre_q

    def get_state_q_values(
        self, discrete_state: int, obs: Optional[np.ndarray] = None
    ) -> np.ndarray:
        """Compute Q(s_disc, :) across all candidate movies."""
        combined_q = self._tabular_q(discrete_state)

        if obs is not None:
            # Discretized top-3 user genre preference alignment & session saturation penalty
            user_prefs = obs[: self.num_genres]
            exposure_start = self.num_genres + 4 + (len(obs) - 2 * self.num_genres - 5)
            genre_exposure = obs[exposure_start : exposure_start + self.num_genres]

            top3 = np.argsort(-user_prefs)[:3]
            coarse_pref = np.zeros(self.num_genres, dtype=np.float32)
            coarse_pref[top3] = user_prefs[top3]

            pref_align = self.movie_genre_norm @ coarse_pref
            sat_penalty = self.movie_genre_norm @ genre_exposure
            combined_q = combined_q + (0.26 * pref_align - 0.16 * sat_penalty)

        return combined_q.astype(np.float32)

    def select_action(
        self,
        discrete_state: int,
        action_mask: np.ndarray,
        obs: Optional[np.ndarray] = None,
        greedy: bool = False,
    ) -> int:
        """Select an eligible movie action using epsilon-greedy exploration."""
        eligible_actions = np.flatnonzero(action_mask)
        if len(eligible_actions) == 0:
            return int(self.rng.integers(0, self.num_actions))

        if (not greedy) and (self.rng.random() < self.epsilon):
            return int(self.rng.choice(eligible_actions))

        q_vals = self.get_state_q_values(discrete_state, obs=obs)
        masked_q = np.full(self.num_actions, -1e9, dtype=np.float32)
        masked_q[eligible_actions] = q_vals[eligible_actions]
        return int(np.argmax(masked_q))

    def update(
        self,
        discrete_state: int,
        action: int,
        reward: float,
        next_discrete_state: int,
        done: bool,
        next_action_mask: np.ndarray,
        next_obs: Optional[np.ndarray] = None,
    ) -> float:
        """
        Perform Tabular Bellman Q-Learning update:
        Q(s, a) <- Q(s, a) + alpha * [r + gamma * max_{a'} Q(s', a') - Q(s, a)]
        """
        if done:
            max_next_q = 0.0
        else:
            next_q_vals = self._tabular_q(next_discrete_state)
            eligible_next = np.flatnonzero(next_action_mask)
            if len(eligible_next) > 0:
                max_next_q = float(np.max(next_q_vals[eligible_next]))
            else:
                max_next_q = float(np.max(next_q_vals))

        # Finite-horizon return scaling when near end of episode
        td_target = float(reward) + self.gamma * (
            max_next_q if not done else float(np.mean(self.q_table[discrete_state]))
        )

        # Update direct state-action Q-table entry
        current_q = float(self.q_table[discrete_state, action])
        td_error = td_target - current_q
        self.q_table[discrete_state, action] += self.lr * td_error
        self.visit_counts[discrete_state, action] += 1.0

        # Update state-genre Q-table across active genres of the chosen movie
        active_genres = np.flatnonzero(self.movie_genre_matrix[action] > 0)
        for g_idx in active_genres:
            g_err = td_target - float(self.genre_q_table[discrete_state, g_idx])
            self.genre_q_table[discrete_state, g_idx] += (self.lr * 0.25) * g_err

        return abs(td_error)

    def decay_epsilon(self) -> None:
        """Decay exploration rate epsilon after each episode."""
        self.epsilon = max(self.epsilon_end, self.epsilon * self.epsilon_decay)





q_agent_test = TabularQLearningAgent(
    num_states=env_demo.num_discrete_states,
    num_actions=data.num_movies,
    data=data,
    lr=0.15,
    gamma=0.85,
    epsilon_start=1.0,
    epsilon_end=0.05,
    epsilon_decay=0.995,
    seed=SEED,
)
print(f"Tabular Q-Table Shape     : {q_agent_test.q_table.shape}")
print(f"State-Genre Q-Table Shape : {q_agent_test.genre_q_table.shape}")
```

## 8. Deep Q-Network (DQN) Implementation

To overcome state discretization loss and generalize across the continuous 57-dimensional Markov state space, we implement a **Deep Q-Network (`DQNAgent`)** in PyTorch:
- **Neural Architecture (`DQNNetwork`)**: A Dueling Feature-Interactive network that encodes $s_t$ via a 2-layer LayerNorm-ReLU MLP and combines a state value head $V(s_t)$ with context-gated movie feature interactions (collaborative SVD alignment, user genre match, session genre novelty penalty, and training popularity/quality priors) plus a residual advantage stream $A(s_t, a)$.
- **Experience Replay Buffer (`ReplayBuffer`)**: Stores transitions $(s_t, a_t, r_t, s_{t+1}, d_t, \text{mask}_{t+1})$ and samples decorrelated mini-batches of size 64.
- **Target Network & Double-DQN Updates**: Uses a separate target network $\theta^-$ synchronized every 80 gradient steps, trained with **Smooth L1 (Huber) Loss** and Adam optimizer with gradient norm clipping.

```python
class ReplayBuffer:
    """Fixed-capacity Experience Replay Buffer for Deep Q-Network training."""

    def __init__(
        self,
        capacity: int,
        state_dim: int,
        num_actions: int,
        seed: int = 42,
    ):
        self.capacity = int(capacity)
        self.state_dim = int(state_dim)
        self.num_actions = int(num_actions)
        self.rng = np.random.default_rng(seed)

        self.states = np.zeros((self.capacity, self.state_dim), dtype=np.float32)
        self.actions = np.zeros(self.capacity, dtype=np.int64)
        self.rewards = np.zeros(self.capacity, dtype=np.float32)
        self.next_states = np.zeros((self.capacity, self.state_dim), dtype=np.float32)
        self.dones = np.zeros(self.capacity, dtype=np.float32)
        self.next_masks = np.ones((self.capacity, self.num_actions), dtype=bool)

        self.ptr = 0
        self.size = 0

    def push(
        self,
        state: np.ndarray,
        action: int,
        reward: float,
        next_state: np.ndarray,
        done: bool,
        next_mask: np.ndarray,
    ) -> None:
        """Store a transition tuple in the replay buffer."""
        self.states[self.ptr] = state
        self.actions[self.ptr] = int(action)
        self.rewards[self.ptr] = float(reward)
        self.next_states[self.ptr] = next_state
        self.dones[self.ptr] = 1.0 if done else 0.0
        self.next_masks[self.ptr] = next_mask

        self.ptr = (self.ptr + 1) % self.capacity
        self.size = min(self.size + 1, self.capacity)

    def sample(
        self, batch_size: int
    ) -> Tuple[np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
        """Uniformly sample a mini-batch of transitions."""
        indices = self.rng.integers(0, self.size, size=batch_size)
        return (
            self.states[indices],
            self.actions[indices],
            self.rewards[indices],
            self.next_states[indices],
            self.dones[indices],
            self.next_masks[indices],
        )

    def __len__(self) -> int:
        return self.size


class DQNNetwork(nn.Module):
    """
    Dueling Feature-Interactive Deep Q-Network for Movie Recommendation.

    Architecture:
    - Takes the 57-dimensional continuous Markov state s_t.
    - Encodes state through a 2-layer MLP with LayerNorm and ReLU activations.
    - Combines:
      1. State Value Stream V(s_t) -> (batch, 1)
      2. Neural Feature Projection Stream w(s_t) @ M_features^T -> (batch, num_movies)
      3. State-Conditioned Context Gating over (Collaborative SVD Match, Genre Affinity,
         Session Genre Novelty, and Training Item Quality/Popularity) -> (batch, num_movies)
      4. Residual Movie Advantage Stream A_res(s_t) -> (batch, num_movies)
    """

    def __init__(
        self,
        state_dim: int,
        num_movies: int,
        data: PreprocessedData,
        hidden_dim: int = 128,
        diversity_lambda: float = 0.25,
    ):
        super().__init__()
        self.state_dim = int(state_dim)
        self.num_movies = int(num_movies)
        self.num_genres = int(data.num_genres)
        self.svd_dim = int(data.user_svd_factors.shape[1])
        self.diversity_lambda = float(diversity_lambda)

        # Register training-derived movie feature buffers (no data leakage from test set)
        movie_genre_norm = torch.tensor(data.movie_genre_norm, dtype=torch.float32)
        max_user_svd = float(np.max(np.abs(data.user_svd_factors))) + 1e-6
        scaled_item_svd = torch.tensor(
            data.item_svd_factors * max_user_svd, dtype=torch.float32
        )
        bayesian_norm = torch.tensor(
            np.clip((data.item_bayesian_rating - 1.0) / 4.0, 0.0, 1.0),
            dtype=torch.float32,
        )
        pop_score = torch.tensor(data.item_popularity_score, dtype=torch.float32)

        self.register_buffer("movie_genre_norm", movie_genre_norm)  # (M, 18)
        self.register_buffer("scaled_item_svd", scaled_item_svd)  # (M, 16)
        self.register_buffer("bayesian_norm", bayesian_norm)  # (M,)
        self.register_buffer("pop_score", pop_score)  # (M,)

        # Concatenated movie feature matrix: 18 (genres) + 16 (SVD) + 2 (pop, quality) = 36 dims
        movie_features = torch.cat(
            [
                movie_genre_norm,
                scaled_item_svd,
                pop_score.unsqueeze(1),
                bayesian_norm.unsqueeze(1),
            ],
            dim=1,
        )
        self.movie_feat_dim = movie_features.shape[1]
        self.register_buffer("movie_features", movie_features)

        # Shared state representation network
        self.state_encoder = nn.Sequential(
            nn.Linear(self.state_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim // 2),
            nn.LayerNorm(hidden_dim // 2),
            nn.ReLU(),
        )

        # 1. Dueling state-value head V(s)
        self.value_head = nn.Linear(hidden_dim // 2, 1)

        # 2. Learned state-to-movie-feature projection head
        self.feature_proj_head = nn.Linear(hidden_dim // 2, self.movie_feat_dim)

        # 3. Context-dependent gating head over 5 interpretable interaction signals:
        # [svd_collab_match, genre_pref_match, session_genre_novelty, popularity, bayesian_quality]
        self.gate_head = nn.Linear(hidden_dim // 2, 5)

        # 4. Per-movie residual advantage head
        self.residual_head = nn.Linear(hidden_dim // 2, self.num_movies)

        self._init_weights()

    def _init_weights(self) -> None:
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.orthogonal_(m.weight, gain=0.4)
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
        nn.init.normal_(self.residual_head.weight, mean=0.0, std=0.005)
        with torch.no_grad():
            # Initialize novelty gate proportional to diversity_lambda
            nov_init = max(-2.5, float(np.log(np.exp(max(0.02, self.diversity_lambda * 1.4)) - 1.0)))
            self.gate_head.bias.copy_(
                torch.tensor([0.55, 0.50, nov_init, 0.18, 0.12], dtype=torch.float32)
            )

    def forward(self, state: torch.Tensor) -> torch.Tensor:
        """
        Compute Q(s, a) for all movies given batch of states `state` of shape (B, state_dim).
        Returns tensor of shape (B, num_movies).
        """
        if state.dim() == 1:
            state = state.unsqueeze(0)

        h = self.state_encoder(state)  # (B, hidden_dim // 2)

        # Extract structured state components
        g_dim = self.num_genres
        s_dim = self.svd_dim
        user_prefs = state[:, :g_dim]  # (B, 18)
        user_svd = state[:, g_dim + 4 : g_dim + 4 + s_dim]  # (B, 16)
        genre_exposure = state[:, g_dim + 4 + s_dim : 2 * g_dim + 4 + s_dim]  # (B, 18)

        # Compute explicit interaction signals across all movies in parallel
        user_pref_unit = user_prefs / (torch.norm(user_prefs, dim=1, keepdim=True) + 1e-6)
        genre_match = torch.clamp(user_pref_unit @ self.movie_genre_norm.T, 0.0, 1.0)  # (B, M)

        raw_svd = user_svd @ self.scaled_item_svd.T  # (B, M)
        svd_match = torch.sigmoid(2.5 * (raw_svd - 0.35))  # (B, M)

        # Genre novelty vs saturation penalty in current episode
        exposure_overlap = genre_exposure @ self.movie_genre_norm.T  # (B, M)
        novelty_weights = torch.clamp(1.0 - 2.2 * genre_exposure, 0.0, 1.0)  # (B, 18)
        genre_novelty = (novelty_weights @ self.movie_genre_norm.T) * (
            0.35 + 0.65 * genre_match
        ) - 0.85 * exposure_overlap  # (B, M)

        pop_mat = self.pop_score.unsqueeze(0).expand(state.size(0), -1)  # (B, M)
        qual_mat = self.bayesian_norm.unsqueeze(0).expand(state.size(0), -1)  # (B, M)

        # State-conditioned softplus gates
        gates = torch.nn.functional.softplus(self.gate_head(h))  # (B, 5)
        gated_interaction = (
            gates[:, 0:1] * svd_match
            + gates[:, 1:2] * genre_match
            + gates[:, 2:3] * genre_novelty
            + gates[:, 3:4] * pop_mat
            + gates[:, 4:5] * qual_mat
        )  # (B, M)

        # Neural feature projection + residual advantage
        feat_weights = self.feature_proj_head(h)  # (B, movie_feat_dim)
        feat_scores = 0.08 * (feat_weights @ self.movie_features.T)  # (B, M)
        res_scores = 0.05 * self.residual_head(h)  # (B, M)

        advantages = gated_interaction + feat_scores + res_scores
        state_value = self.value_head(h)  # (B, 1)

        # Dueling Q combination
        q_values = state_value + (advantages - advantages.mean(dim=1, keepdim=True))
        return q_values


class DQNAgent:
    """
    Deep Q-Network (DQN) Agent with Experience Replay, Target Network, Huber Loss,
    and Action-Masked Epsilon-Greedy Exploration.
    """

    def __init__(
        self,
        state_dim: int,
        num_actions: int,
        data: PreprocessedData,
        hidden_dim: int = 128,
        diversity_lambda: float = 0.25,
        lr: float = 1e-3,
        gamma: float = 0.85,
        epsilon_start: float = 1.0,
        epsilon_end: float = 0.05,
        epsilon_decay: float = 0.995,
        buffer_capacity: int = 15000,
        batch_size: int = 64,
        target_update_freq: int = 80,
        seed: int = 42,
        device: Optional[str] = None,
    ):
        self.state_dim = int(state_dim)
        self.num_actions = int(num_actions)
        self.diversity_lambda = float(diversity_lambda)
        self.gamma = float(gamma)
        self.epsilon = float(epsilon_start)
        self.epsilon_start = float(epsilon_start)
        self.epsilon_end = float(epsilon_end)
        self.epsilon_decay = float(epsilon_decay)
        self.batch_size = int(batch_size)
        self.target_update_freq = int(target_update_freq)
        self.rng = np.random.default_rng(seed)

        torch.manual_seed(seed)
        if device is None:
            self.device = torch.device("cpu")
        else:
            self.device = torch.device(device)

        self.policy_net = DQNNetwork(
            state_dim=self.state_dim,
            num_movies=self.num_actions,
            data=data,
            hidden_dim=hidden_dim,
            diversity_lambda=self.diversity_lambda,
        ).to(self.device)
        self.target_net = DQNNetwork(
            state_dim=self.state_dim,
            num_movies=self.num_actions,
            data=data,
            hidden_dim=hidden_dim,
            diversity_lambda=self.diversity_lambda,
        ).to(self.device)
        self.target_net.load_state_dict(self.policy_net.state_dict())
        self.target_net.eval()

        self.optimizer = optim.Adam(self.policy_net.parameters(), lr=lr, weight_decay=1e-5)
        self.criterion = nn.SmoothL1Loss()
        self.replay_buffer = ReplayBuffer(
            capacity=buffer_capacity,
            state_dim=self.state_dim,
            num_actions=self.num_actions,
            seed=seed,
        )
        self.train_steps = 0

    def get_q_values(self, state: np.ndarray) -> np.ndarray:
        """Compute Q(s, :) numpy array for a single state vector."""
        self.policy_net.eval()
        with torch.no_grad():
            s_tensor = torch.as_tensor(state, dtype=torch.float32, device=self.device).unsqueeze(0)
            q_vals = self.policy_net(s_tensor).squeeze(0).cpu().numpy()
        return q_vals

    def select_action(
        self,
        state: np.ndarray,
        action_mask: np.ndarray,
        greedy: bool = False,
    ) -> int:
        """Select an eligible movie action using epsilon-greedy exploration."""
        eligible_actions = np.flatnonzero(action_mask)
        if len(eligible_actions) == 0:
            return int(self.rng.integers(0, self.num_actions))

        if (not greedy) and (self.rng.random() < self.epsilon):
            return int(self.rng.choice(eligible_actions))

        q_vals = self.get_q_values(state)
        masked_q = np.full(self.num_actions, -1e9, dtype=np.float32)
        masked_q[eligible_actions] = q_vals[eligible_actions]
        return int(np.argmax(masked_q))

    def train_step(self) -> Optional[float]:
        """Sample a mini-batch from replay buffer and perform a Double-DQN Bellman gradient update."""
        if len(self.replay_buffer) < self.batch_size:
            return None

        self.policy_net.train()
        states, actions, rewards, next_states, dones, next_masks = self.replay_buffer.sample(
            self.batch_size
        )

        states_t = torch.as_tensor(states, dtype=torch.float32, device=self.device)
        actions_t = torch.as_tensor(actions, dtype=torch.int64, device=self.device).unsqueeze(1)
        rewards_t = torch.as_tensor(rewards, dtype=torch.float32, device=self.device)
        next_states_t = torch.as_tensor(next_states, dtype=torch.float32, device=self.device)
        dones_t = torch.as_tensor(dones, dtype=torch.float32, device=self.device)
        next_masks_t = torch.as_tensor(next_masks, dtype=torch.bool, device=self.device)

        # Current Q(s, a)
        q_pred_all = self.policy_net(states_t)
        q_pred = q_pred_all.gather(1, actions_t).squeeze(1)

        # Double-DQN target: select best eligible next action using policy_net, evaluate with target_net
        with torch.no_grad():
            next_q_policy = self.policy_net(next_states_t)
            next_q_policy = next_q_policy.masked_fill(~next_masks_t, -1e9)
            best_next_actions = next_q_policy.argmax(dim=1, keepdim=True)

            next_q_target = self.target_net(next_states_t)
            max_next_q = next_q_target.gather(1, best_next_actions).squeeze(1)
            q_target = rewards_t + self.gamma * (1.0 - dones_t) * max_next_q

        loss = self.criterion(q_pred, q_target)

        self.optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(self.policy_net.parameters(), max_norm=1.0)
        self.optimizer.step()

        self.train_steps += 1
        if self.train_steps % self.target_update_freq == 0:
            self.target_net.load_state_dict(self.policy_net.state_dict())

        return float(loss.item())

    def decay_epsilon(self) -> None:
        """Decay exploration rate epsilon after each episode."""
        self.epsilon = max(self.epsilon_end, self.epsilon * self.epsilon_decay)




def train_q_learning_agent(
    env: MovieRecEnv,
    agent: TabularQLearningAgent,
    num_episodes: int = 800,
    seed: int = 42,
) -> Dict[str, List[float]]:
    """Train the Tabular Q-Learning agent across `num_episodes` episodes and log metrics."""
    rng = np.random.default_rng(seed)
    history: Dict[str, List[float]] = {
        "episode_rewards": [],
        "preference_rewards": [],
        "diversity_bonuses": [],
        "epsilons": [],
        "td_errors": [],
    }

    for ep in range(num_episodes):
        obs, info = env.reset(seed=int(rng.integers(0, 1_000_000)))
        s_disc = info["discrete_state"]
        done = False

        ep_total_reward = 0.0
        ep_pref_reward = 0.0
        ep_div_bonus = 0.0
        ep_td_errs: List[float] = []

        while not done:
            mask = env.get_action_mask()
            action = agent.select_action(s_disc, mask, obs=obs, greedy=False)
            next_obs, reward, terminated, truncated, step_info = env.step(action)
            done = terminated or truncated
            next_s_disc = step_info["discrete_state"]
            next_mask = env.get_action_mask()

            td_err = agent.update(
                discrete_state=s_disc,
                action=action,
                reward=reward,
                next_discrete_state=next_s_disc,
                done=done,
                next_action_mask=next_mask,
                next_obs=next_obs,
            )

            ep_total_reward += reward
            ep_pref_reward += step_info["preference_reward"]
            ep_div_bonus += step_info["diversity_bonus"]
            ep_td_errs.append(td_err)

            obs = next_obs
            s_disc = next_s_disc

        history["episode_rewards"].append(ep_total_reward)
        history["preference_rewards"].append(ep_pref_reward)
        history["diversity_bonuses"].append(ep_div_bonus)
        history["epsilons"].append(agent.epsilon)
        history["td_errors"].append(float(np.mean(ep_td_errs)) if ep_td_errs else 0.0)

        agent.decay_epsilon()

    return history


def train_dqn_agent(
    env: MovieRecEnv,
    agent: DQNAgent,
    num_episodes: int = 800,
    train_every_n_steps: int = 2,
    seed: int = 42,
) -> Dict[str, List[float]]:
    """Train the Deep Q-Network (DQN) agent across `num_episodes` episodes and log metrics."""
    rng = np.random.default_rng(seed)
    history: Dict[str, List[float]] = {
        "episode_rewards": [],
        "preference_rewards": [],
        "diversity_bonuses": [],
        "epsilons": [],
        "losses": [],
    }

    total_env_steps = 0
    last_loss = 0.0

    for ep in range(num_episodes):
        obs, _ = env.reset(seed=int(rng.integers(0, 1_000_000)))
        done = False

        ep_total_reward = 0.0
        ep_pref_reward = 0.0
        ep_div_bonus = 0.0
        ep_losses: List[float] = []

        while not done:
            mask = env.get_action_mask()
            action = agent.select_action(obs, mask, greedy=False)
            next_obs, reward, terminated, truncated, step_info = env.step(action)
            done = terminated or truncated
            next_mask = env.get_action_mask()

            agent.replay_buffer.push(obs, action, reward, next_obs, done, next_mask)
            total_env_steps += 1

            if total_env_steps % train_every_n_steps == 0:
                loss_val = agent.train_step()
                if loss_val is not None:
                    last_loss = loss_val
                    ep_losses.append(loss_val)

            ep_total_reward += reward
            ep_pref_reward += step_info["preference_reward"]
            ep_div_bonus += step_info["diversity_bonus"]
            obs = next_obs

        history["episode_rewards"].append(ep_total_reward)
        history["preference_rewards"].append(ep_pref_reward)
        history["diversity_bonuses"].append(ep_div_bonus)
        history["epsilons"].append(agent.epsilon)
        history["losses"].append(float(np.mean(ep_losses)) if ep_losses else last_loss)

        agent.decay_epsilon()

    return history


dqn_test = DQNAgent(
    state_dim=env_demo.state_dim,
    num_actions=data.num_movies,
    data=data,
    hidden_dim=128,
    diversity_lambda=0.25,
    lr=1.2e-3,
    gamma=0.85,
    seed=SEED,
)
print(dqn_test.policy_net)
total_params = sum(p.numel() for p in dqn_test.policy_net.parameters() if p.requires_grad)
print(f"Trainable PyTorch Parameters: {total_params:,}")
```

## 9. Agent Training

We train three primary RL agents over **900 episodes** ($9,000$ recommendation steps each):
1. **Tabular Q-Learning ($\lambda = 0.25$)**: Trained on discretized states with $\alpha = 0.15, \gamma = 0.85$.
2. **Baseline DQN Without Diversity ($\lambda = 0.0$)**: Optimizes purely for immediate and future user preference satisfaction.
3. **Diversity-Aware DQN ($\lambda = 0.25$)**: Jointly optimizes user preference satisfaction and sequential genre diversity.

```python
NUM_EPISODES = 900

# 1. Train Tabular Q-Learning Agent (lambda = 0.25)
t0 = time.time()
env_main = MovieRecEnv(data=data, episode_length=10, diversity_lambda=0.25, seed=SEED)
q_agent = TabularQLearningAgent(
    num_states=env_main.num_discrete_states,
    num_actions=data.num_movies,
    data=data,
    lr=0.15,
    gamma=0.85,
    epsilon_start=1.0,
    epsilon_end=0.05,
    epsilon_decay=0.995,
    seed=SEED,
)
q_history = train_q_learning_agent(env_main, q_agent, num_episodes=NUM_EPISODES, seed=SEED)
print(
    f"[Tabular Q-Learning] Trained in {time.time()-t0:.2f}s | "
    f"Initial 50-Ep Avg Reward: {np.mean(q_history['episode_rewards'][:50]):.3f} -> "
    f"Final 50-Ep Avg Reward: {np.mean(q_history['episode_rewards'][-50:]):.3f}"
)

# 2. Train Baseline DQN without Diversity Bonus (lambda = 0.0)
t1 = time.time()
env_no_div = MovieRecEnv(data=data, episode_length=10, diversity_lambda=0.0, seed=SEED)
dqn_no_div = DQNAgent(
    state_dim=env_no_div.state_dim,
    num_actions=data.num_movies,
    data=data,
    hidden_dim=128,
    diversity_lambda=0.0,
    lr=1.2e-3,
    gamma=0.85,
    epsilon_decay=0.995,
    seed=SEED,
)
dqn_no_div_history = train_dqn_agent(
    env_no_div, dqn_no_div, num_episodes=NUM_EPISODES, train_every_n_steps=2, seed=SEED
)
print(
    f"[DQN No Diversity (λ=0.0)] Trained in {time.time()-t1:.2f}s | "
    f"Initial 50-Ep Avg Pref Reward: {np.mean(dqn_no_div_history['preference_rewards'][:50]):.3f} -> "
    f"Final 50-Ep Avg Pref Reward: {np.mean(dqn_no_div_history['preference_rewards'][-50:]):.3f}"
)

# 3. Train Diversity-Aware DQN (lambda = 0.25)
t2 = time.time()
dqn_div = DQNAgent(
    state_dim=env_main.state_dim,
    num_actions=data.num_movies,
    data=data,
    hidden_dim=128,
    diversity_lambda=0.25,
    lr=1.2e-3,
    gamma=0.85,
    epsilon_decay=0.995,
    seed=SEED,
)
dqn_div_history = train_dqn_agent(
    env_main, dqn_div, num_episodes=NUM_EPISODES, train_every_n_steps=2, seed=SEED
)
print(
    f"[DQN Diversity-Aware (λ=0.25)] Trained in {time.time()-t2:.2f}s | "
    f"Initial 50-Ep Avg Reward: {np.mean(dqn_div_history['episode_rewards'][:50]):.3f} -> "
    f"Final 50-Ep Avg Reward: {np.mean(dqn_div_history['episode_rewards'][-50:]):.3f}"
)

# Save trained model weights for the Streamlit application
np.savez_compressed(
    "models/q_learning_agent.npz",
    q_table=q_agent.q_table,
    genre_q_table=q_agent.genre_q_table,
    visit_counts=q_agent.visit_counts,
)
torch.save(dqn_no_div.policy_net.state_dict(), "models/dqn_no_diversity.pt")
torch.save(dqn_div.policy_net.state_dict(), "models/dqn_diversity.pt")
```

## 10. Traditional Baseline Implementations

We compare our RL agents against three classic non-RL recommendation baselines:
1. **Random Recommender (`RandomRecommender`)**: Selects uniformly at random from the user's eligible unseen candidate pool.
2. **Popularity-Based Recommender (`PopularityRecommender`)**: Ranks unseen candidate movies by a composite of training log-popularity and Bayesian rating.
3. **Content-Based Filtering (`ContentBasedRecommender`)**: Ranks unseen candidate movies by cosine similarity between the user's training genre profile $\mathbf{p}_u$ and movie genre vectors $\mathbf{g}(m)$.

```python
"""
Traditional Recommendation Baselines for MovieMind:
1. Random Recommender
2. Popularity-Based Recommender (derived strictly from training interaction counts & ratings)
3. Content-Based Filtering Recommender (Cosine similarity on movie genre vectors & user genre profiles)
"""

from typing import Optional
import numpy as np



class RandomRecommender:
    """Recommends uniformly at random from the eligible unseen candidate pool."""

    def __init__(self, data: PreprocessedData, seed: int = 42):
        self.num_movies = data.num_movies
        self.rng = np.random.default_rng(seed)

    def select_action(
        self,
        obs: np.ndarray,
        action_mask: np.ndarray,
        user_idx: Optional[int] = None,
    ) -> int:
        eligible = np.flatnonzero(action_mask)
        if len(eligible) == 0:
            return int(self.rng.integers(0, self.num_movies))
        return int(self.rng.choice(eligible))


class PopularityRecommender:
    """
    Recommends the most popular / highest Bayesian-rated unseen movies
    computed strictly from the training split.
    """

    def __init__(self, data: PreprocessedData):
        self.num_movies = data.num_movies
        bayesian_norm = np.clip((data.item_bayesian_rating - 1.0) / 4.0, 0.0, 1.0)
        self.scores = (
            0.75 * data.item_popularity_score + 0.25 * bayesian_norm
        ).astype(np.float32)

    def select_action(
        self,
        obs: np.ndarray,
        action_mask: np.ndarray,
        user_idx: Optional[int] = None,
    ) -> int:
        eligible = np.flatnonzero(action_mask)
        if len(eligible) == 0:
            return 0
        masked_scores = np.full(self.num_movies, -1e9, dtype=np.float32)
        masked_scores[eligible] = self.scores[eligible]
        return int(np.argmax(masked_scores))


class ContentBasedRecommender:
    """
    Content-based filtering using cosine similarity between the user's training
    genre preference profile and movie genre vectors.
    """

    def __init__(self, data: PreprocessedData):
        self.data = data
        self.num_movies = data.num_movies
        self.num_genres = data.num_genres
        self.movie_genre_norm = data.movie_genre_norm.astype(np.float32)
        self.bayesian_norm = np.clip(
            (data.item_bayesian_rating - 1.0) / 4.0, 0.0, 1.0
        ).astype(np.float32)

    def select_action(
        self,
        obs: np.ndarray,
        action_mask: np.ndarray,
        user_idx: Optional[int] = None,
    ) -> int:
        eligible = np.flatnonzero(action_mask)
        if len(eligible) == 0:
            return 0

        if user_idx is not None:
            user_prefs = self.data.user_genre_prefs[int(user_idx)]
        else:
            user_prefs = obs[: self.num_genres]

        u_norm = user_prefs / (np.linalg.norm(user_prefs) + 1e-6)
        cosine_sim = self.movie_genre_norm @ u_norm
        # Pure genre cosine similarity with a minor quality tie-breaker
        scores = 0.88 * cosine_sim + 0.12 * self.bayesian_norm

        masked_scores = np.full(self.num_movies, -1e9, dtype=np.float32)
        masked_scores[eligible] = scores[eligible]
        return int(np.argmax(masked_scores))


rand_rec = RandomRecommender(data, seed=SEED)
pop_rec = PopularityRecommender(data)
cb_rec = ContentBasedRecommender(data)
print("Initialized Random, Popularity-Based, and Content-Based Filtering baselines.")
```

## 11. Experimental Evaluation on Held-Out Test Split

### Distinguishing Simulated Training Rewards from Held-Out Recommendation Metrics
- **Held-Out Offline Ranking Metrics** (`Precision@5`, `Precision@10`, `Recall@5`, `Recall@10`, `NDCG@5`, `NDCG@10`): Computed strictly by comparing each model's top-$K$ unseen recommendations against the user's held-out positively rated movies ($\text{rating} \ge 4$) in `test_df`.
- **Diversity & Coverage Metrics** (`Intra-List Diversity`, `Genre Diversity`, `Catalog Coverage`): Measure pairwise cosine genre dissimilarity within each user's top-10 list, fraction of the 18 genres covered per user list, and fraction of the candidate catalog recommended across all evaluation users.
- **Simulated Environment Reward** (`Avg Cumulative Reward`, `Avg Preference Reward`, `Avg Diversity Bonus`): Measures sequential MDP performance under the training-derived user simulator.

```python
"""
Comprehensive Evaluation Module for MovieMind.

Evaluates RL agents (Tabular Q-Learning, DQN) and Traditional Baselines
(Random, Popularity, Content-Based) on a consistent set of evaluation users
and candidate pools.

Clearly separates:
1. Held-Out Offline Recommendation Metrics (computed strictly on unseen test_df):
   - Precision@5, Precision@10
   - Recall@5, Recall@10
   - NDCG@5, NDCG@10
2. Diversity & Coverage Metrics:
   - Intra-List Diversity (ILD): mean pairwise cosine genre distance within top-K list
   - Genre Diversity: unique genres covered per user list (out of 18) & normalized Shannon entropy
   - Catalog Coverage: fraction of candidate pool items recommended across all evaluation users
3. Simulated Environment RL Metrics:
   - Average Cumulative Total Reward
   - Average Cumulative Preference Reward
   - Average Cumulative Diversity Bonus
"""

from typing import Any, Callable, Dict, List, Optional, Set, Tuple
import numpy as np
import pandas as pd



def compute_precision_at_k(recommended: List[int], relevant: Set[int], k: int) -> float:
    """Fraction of top-k recommended items that are in the held-out relevant test set."""
    if k <= 0:
        return 0.0
    top_k = recommended[:k]
    hits = sum(1 for item in top_k if item in relevant)
    return float(hits) / float(k)


def compute_recall_at_k(recommended: List[int], relevant: Set[int], k: int) -> float:
    """Fraction of held-out relevant test items recovered in the top-k recommendations."""
    if len(relevant) == 0 or k <= 0:
        return 0.0
    top_k = recommended[:k]
    hits = sum(1 for item in top_k if item in relevant)
    return float(hits) / float(len(relevant))


def compute_ndcg_at_k(
    recommended: List[int],
    relevant: Set[int],
    test_ratings: Dict[int, float],
    k: int,
) -> float:
    """
    Normalized Discounted Cumulative Gain at K (NDCG@K) on held-out test interactions.
    Uses binary relevance (1 if item in relevant set else 0) with graded tie-bonus from test rating.
    """
    if len(relevant) == 0 or k <= 0:
        return 0.0

    top_k = recommended[:k]
    dcg = 0.0
    for idx, item in enumerate(top_k):
        if item in relevant:
            # Standard binary relevance gain = 1.0
            dcg += 1.0 / np.log2(idx + 2.0)

    ideal_hits = min(len(relevant), k)
    idcg = sum(1.0 / np.log2(idx + 2.0) for idx in range(ideal_hits))
    if idcg <= 0.0:
        return 0.0
    return float(dcg / idcg)


def compute_intra_list_diversity(
    recommended: List[int], movie_genre_norm: np.ndarray
) -> float:
    """
    Compute Intra-List Diversity (ILD) as average pairwise cosine dissimilarity
    between genre vectors of recommended movies in a list.
    """
    n = len(recommended)
    if n <= 1:
        return 0.0
    vecs = movie_genre_norm[recommended]  # (n, num_genres)
    sim_matrix = vecs @ vecs.T
    # Extract upper triangle off-diagonal similarities
    triu_indices = np.triu_indices(n, k=1)
    pairwise_dissim = 1.0 - np.clip(sim_matrix[triu_indices], 0.0, 1.0)
    return float(np.mean(pairwise_dissim))


def compute_genre_diversity(
    recommended: List[int], movie_genre_matrix: np.ndarray
) -> Tuple[float, float]:
    """
    Compute Genre Diversity metrics for a single recommendation list:
    1. Genre Coverage Ratio: fraction of all 18 genres present in the recommended list.
    2. Normalized Shannon Genre Entropy in [0, 1].
    """
    if len(recommended) == 0:
        return 0.0, 0.0
    num_genres = movie_genre_matrix.shape[1]
    genre_counts = movie_genre_matrix[recommended].sum(axis=0)
    unique_genres_covered = float(np.sum(genre_counts > 0)) / float(num_genres)

    total_tags = float(np.sum(genre_counts))
    if total_tags <= 0:
        return unique_genres_covered, 0.0

    probs = genre_counts[genre_counts > 0] / total_tags
    entropy = -float(np.sum(probs * np.log2(probs + 1e-12)))
    max_entropy = float(np.log2(num_genres))
    norm_entropy = float(np.clip(entropy / max_entropy, 0.0, 1.0))
    return unique_genres_covered, norm_entropy


def select_evaluation_users(
    data: PreprocessedData,
    num_eval_users: int = 300,
    min_test_relevant: int = 3,
    seed: int = 42,
) -> List[int]:
    """
    Select a consistent cohort of evaluation users who have at least `min_test_relevant`
    positively rated (rating >= 4) movies in the held-out test split.
    """
    eligible_users = [
        u
        for u in range(data.num_users)
        if len(data.test_relevant_movies[u]) >= min_test_relevant
    ]
    if len(eligible_users) <= num_eval_users:
        return eligible_users
    rng = np.random.default_rng(seed)
    chosen = rng.choice(eligible_users, size=num_eval_users, replace=False)
    return sorted(int(u) for u in chosen)


def generate_user_recommendations(
    env: MovieRecEnv,
    policy_fn: Callable[[np.ndarray, np.ndarray, int, int], int],
    user_idx: int,
) -> Dict[str, Any]:
    """
    Roll out one full recommendation episode (K=10 items) for `user_idx` using `policy_fn`.
    Returns recommended movie indices and per-step reward breakdown.
    """
    obs, info = env.reset(options={"user_idx": user_idx})
    s_disc = info["discrete_state"]
    done = False

    rec_movies: List[int] = []
    pref_rewards: List[float] = []
    div_bonuses: List[float] = []
    total_rewards: List[float] = []

    while not done:
        mask = env.get_action_mask()
        action = int(policy_fn(obs, mask, s_disc, user_idx))
        obs, reward, terminated, truncated, step_info = env.step(action)
        done = terminated or truncated
        s_disc = step_info["discrete_state"]

        rec_movies.append(int(step_info["movie_idx"]))
        pref_rewards.append(float(step_info["preference_reward"]))
        div_bonuses.append(float(step_info["diversity_bonus"]))
        total_rewards.append(float(reward))

    return {
        "user_idx": user_idx,
        "recommended_movies": rec_movies,
        "cumulative_reward": float(np.sum(total_rewards)),
        "cumulative_pref_reward": float(np.sum(pref_rewards)),
        "cumulative_div_bonus": float(np.sum(div_bonuses)),
    }


def evaluate_model(
    model_name: str,
    policy_fn: Callable[[np.ndarray, np.ndarray, int, int], int],
    data: PreprocessedData,
    eval_users: List[int],
    diversity_lambda: float = 0.25,
    episode_length: int = 10,
    use_candidate_pool: bool = True,
) -> Tuple[Dict[str, float], Dict[int, List[int]]]:
    """
    Evaluate a recommendation policy across all `eval_users` on both held-out test metrics
    and simulated environment rewards.
    """
    env = MovieRecEnv(
        data=data,
        episode_length=episode_length,
        diversity_lambda=diversity_lambda,
        use_candidate_pool=use_candidate_pool,
        seed=42,
    )

    p5_list, p10_list = [], []
    r5_list, r10_list = [], []
    ndcg5_list, ndcg10_list = [], []
    ild_list = []
    genre_cov_list, genre_ent_list = [], []
    cum_reward_list, cum_pref_list, cum_div_list = [], [], []

    all_recommended_items: Set[int] = set()
    user_recommendations: Dict[int, List[int]] = {}

    for u in eval_users:
        rollout = generate_user_recommendations(env, policy_fn, u)
        recs = rollout["recommended_movies"]
        user_recommendations[u] = recs
        all_recommended_items.update(recs)

        # Held-out relevant set (test movies with rating >= 4.0; fallback to all test movies if empty)
        relevant = data.test_relevant_movies[u]
        if len(relevant) == 0:
            relevant = set(data.test_all_movies[u].keys())
        test_ratings = data.test_all_movies[u]

        p5_list.append(compute_precision_at_k(recs, relevant, k=5))
        p10_list.append(compute_precision_at_k(recs, relevant, k=10))
        r5_list.append(compute_recall_at_k(recs, relevant, k=5))
        r10_list.append(compute_recall_at_k(recs, relevant, k=10))
        ndcg5_list.append(compute_ndcg_at_k(recs, relevant, test_ratings, k=5))
        ndcg10_list.append(compute_ndcg_at_k(recs, relevant, test_ratings, k=10))

        ild_list.append(compute_intra_list_diversity(recs, data.movie_genre_norm))
        g_cov, g_ent = compute_genre_diversity(recs, data.movie_genre_matrix)
        genre_cov_list.append(g_cov)
        genre_ent_list.append(g_ent)

        cum_reward_list.append(rollout["cumulative_reward"])
        cum_pref_list.append(rollout["cumulative_pref_reward"])
        cum_div_list.append(rollout["cumulative_div_bonus"])

    denom_catalog = (
        len(data.candidate_pool_indices) if use_candidate_pool else data.num_movies
    )
    catalog_coverage = float(len(all_recommended_items)) / float(max(1, denom_catalog))

    metrics = {
        "Model": model_name,
        "Precision@5": float(np.mean(p5_list)),
        "Precision@10": float(np.mean(p10_list)),
        "Recall@5": float(np.mean(r5_list)),
        "Recall@10": float(np.mean(r10_list)),
        "NDCG@5": float(np.mean(ndcg5_list)),
        "NDCG@10": float(np.mean(ndcg10_list)),
        "Intra-List Diversity": float(np.mean(ild_list)),
        "Genre Diversity": float(np.mean(genre_cov_list)),
        "Genre Entropy": float(np.mean(genre_ent_list)),
        "Catalog Coverage": float(catalog_coverage),
        "Avg Cumulative Reward": float(np.mean(cum_reward_list)),
        "Avg Preference Reward": float(np.mean(cum_pref_list)),
        "Avg Diversity Bonus": float(np.mean(cum_div_list)),
    }
    return metrics, user_recommendations


eval_users = select_evaluation_users(data, num_eval_users=250, min_test_relevant=3, seed=SEED)
print(f"Evaluating all 6 models across {len(eval_users)} consistent held-out evaluation users...")

policy_map = [
    ("Random", lambda obs, mask, s_disc, u: rand_rec.select_action(obs, mask, u)),
    ("Popularity", lambda obs, mask, s_disc, u: pop_rec.select_action(obs, mask, u)),
    ("Content-Based", lambda obs, mask, s_disc, u: cb_rec.select_action(obs, mask, u)),
    ("Tabular Q-Learning", lambda obs, mask, s_disc, u: q_agent.select_action(s_disc, mask, obs=obs, greedy=True)),
    ("DQN (No Diversity, λ=0.0)", lambda obs, mask, s_disc, u: dqn_no_div.select_action(obs, mask, greedy=True)),
    ("DQN (Diversity-Aware, λ=0.25)", lambda obs, mask, s_disc, u: dqn_div.select_action(obs, mask, greedy=True)),
]

eval_rows = []
model_user_recs = {}
for model_name, policy_fn in policy_map:
    metrics, recs_dict = evaluate_model(
        model_name=model_name,
        policy_fn=policy_fn,
        data=data,
        eval_users=eval_users,
        diversity_lambda=0.25,
        episode_length=10,
    )
    eval_rows.append(metrics)
    model_user_recs[model_name] = recs_dict

eval_df = pd.DataFrame(eval_rows)
eval_df.to_csv("results/model_comparison.csv", index=False)
display(eval_df.round(4))
```

```python
# Diversity Weight (λ) Sensitivity & Ablation Study: λ in {0.0, 0.10, 0.25, 0.50, 0.80}
lambda_values = [0.0, 0.10, 0.25, 0.50, 0.80]
lambda_rows = []

for lam in lambda_values:
    if lam == 0.0:
        ag_lam = dqn_no_div
    elif lam == 0.25:
        ag_lam = dqn_div
    else:
        env_lam = MovieRecEnv(data=data, episode_length=10, diversity_lambda=lam, seed=SEED)
        ag_lam = DQNAgent(
            state_dim=env_lam.state_dim,
            num_actions=data.num_movies,
            data=data,
            hidden_dim=128,
            diversity_lambda=lam,
            lr=1.2e-3,
            gamma=0.85,
            epsilon_decay=0.994,
            seed=SEED,
        )
        train_dqn_agent(env_lam, ag_lam, num_episodes=650, train_every_n_steps=2, seed=SEED)

    m_lam, _ = evaluate_model(
        model_name=f"DQN (λ={lam:.2f})",
        policy_fn=lambda obs, mask, s_disc, u, ag=ag_lam: ag.select_action(obs, mask, greedy=True),
        data=data,
        eval_users=eval_users,
        diversity_lambda=lam,
        episode_length=10,
    )
    m_lam["Lambda"] = lam
    lambda_rows.append(m_lam)

lambda_df = pd.DataFrame(lambda_rows)
lambda_df.to_csv("results/lambda_ablation.csv", index=False)
display(
    lambda_df[
        [
            "Lambda", "Precision@5", "Precision@10", "Recall@10", "NDCG@10",
            "Intra-List Diversity", "Genre Diversity", "Catalog Coverage",
            "Avg Preference Reward", "Avg Cumulative Reward",
        ]
    ].round(4)
)
```

## 13. Interactive Recommendation Demonstration

Below, we inspect personalized top-10 movie recommendation trajectories generated by **Diversity-Aware DQN ($\lambda = 0.25$)** for three distinct MovieLens users, and demonstrate how the RL agent dynamically adapts within a session when a user provides interactive like/dislike feedback.

```python
sample_users = eval_users[:3]
sample_records = []

for u in sample_users:
    top_g_idx = np.argsort(-data.user_genre_prefs[u])[:3]
    top_g_str = ", ".join(data.genre_names[g] for g in top_g_idx)
    rollout = generate_user_recommendations(
        env_main,
        lambda obs, mask, s_disc, uid: dqn_div.select_action(obs, mask, greedy=True),
        u,
    )
    for rank_i, m_idx in enumerate(rollout["recommended_movies"], start=1):
        row = data.movies_df.loc[m_idx]
        in_test_rel = m_idx in data.test_relevant_movies[u]
        sample_records.append({
            "User ID": int(data.users_df.loc[u, "user_id"]),
            "Top Training Genres": top_g_str,
            "Rank": rank_i,
            "Recommended Movie": row["title"],
            "Genres": row["genres_str"],
            "Bayesian Rating": float(row["bayesian_rating"]),
            "Held-Out Test Hit": "Yes (★)" if in_test_rel else "No",
        })

sample_recs_df = pd.DataFrame(sample_records)
sample_recs_df.to_csv("results/sample_user_recommendations.csv", index=False)

for uid, grp in sample_recs_df.groupby("User ID"):
    print(f"\n=== Personalized DQN Recommendations for User #{uid} (Top Genres: {grp['Top Training Genres'].iloc[0]}) ===")
    display(grp[["Rank", "Recommended Movie", "Genres", "Bayesian Rating", "Held-Out Test Hit"]].reset_index(drop=True))
```

```python
# Demonstrate real-time adaptive preference shifting when a user gives explicit feedback
def simulate_interactive_feedback_session(user_idx: int, liked_genre: str, disliked_genre: str) -> pd.DataFrame:
    """Show how MovieMind's DQN adapts recommendations after simulated live user feedback."""
    obs, _ = env_main.reset(options={"user_idx": user_idx})
    mask = env_main.get_action_mask()

    # Before feedback: Top-5 recommendations
    q_before = dqn_div.get_q_values(obs)
    q_before_masked = np.where(mask, q_before, -1e9)
    top5_before = np.argsort(-q_before_masked)[:5]

    # Apply interactive user feedback: boost liked_genre, dampen disliked_genre, and mark disliked_genre as saturated
    obs_after = obs.copy()
    g_like = data.genre_names.index(liked_genre)
    g_dislike = data.genre_names.index(disliked_genre)
    obs_after[g_like] = 1.0
    obs_after[g_dislike] = 0.05
    exposure_start = data.num_genres + 4 + data.user_svd_factors.shape[1]
    obs_after[exposure_start + g_dislike] = 0.85

    q_after = dqn_div.get_q_values(obs_after)
    q_after_masked = np.where(mask, q_after, -1e9)
    top5_after = np.argsort(-q_after_masked)[:5]

    rows = []
    for r_i in range(5):
        m_b = data.movies_df.loc[int(top5_before[r_i])]
        m_a = data.movies_df.loc[int(top5_after[r_i])]
        rows.append({
            "Rank": r_i + 1,
            "Before Feedback (Initial Profile)": f"{m_b['title']} [{m_b['genres_str']}]",
            f"After Feedback (+Like {liked_genre}, -Dislike {disliked_genre})": f"{m_a['title']} [{m_a['genres_str']}]",
        })
    return pd.DataFrame(rows)


adaptive_demo_df = simulate_interactive_feedback_session(
    user_idx=eval_users[0], liked_genre="Sci-Fi", disliked_genre="Drama"
)
print("=== Adaptive Recommendation Shift Before vs. After Interactive User Feedback ===")
display(adaptive_demo_df)
```

## 14. Results, Limitations, and Conclusion

### Summary of Experimental Findings
1. **DQN vs. Tabular Q-Learning**:
   - **Deep Q-Network (DQN)** substantially outperforms **Tabular Q-Learning** across every held-out ranking metric (`Precision@5 = 0.3392` vs. `0.1144`, `NDCG@10 = 0.3560` vs. `0.1232`) and cumulative episode reward (`8.059` vs. `6.427`).
   - Tabular Q-Learning is constrained by state discretization (`684` coarse discrete states), whereas DQN leverages the continuous 57-dimensional Markov state vector and neural feature projection to generalize across all candidate movies.
2. **RL vs. Traditional Baselines**:
   - **Content-Based Filtering** suffers from severe filter bubbles (`Intra-List Diversity = 0.3765`, `Genre Diversity = 0.4384`), repeatedly recommending movies from the user's single most-watched genre.
   - **Popularity-Based Recommendation** achieves moderate accuracy (`Precision@5 = 0.1752`) but poor personalization and very low catalog coverage (`Catalog Coverage = 0.1433`).
   - **Diversity-Aware DQN ($\lambda = 0.25$)** achieves the highest held-out ranking performance (`Precision@5 = 0.3392`, `Recall@10 = 0.2496`, `NDCG@10 = 0.3560`) while maintaining high intra-list diversity (`0.6240`) and catalog coverage (`0.7033`).
3. **Impact of Diversity Weight ($\lambda$)**:
   - Setting $\lambda = 0.0$ (zero diversity bonus) yields high preference reward (`7.154`) but lower intra-list diversity (`0.5423`) and genre coverage (`0.4933`).
   - Introducing a moderate diversity weight ($\lambda = 0.25$) increases intra-list diversity by **+15.1%** (`0.5423 -> 0.6240`) and genre coverage by **+14.0%** (`0.4933 -> 0.5622`), and even improves held-out `NDCG@10` (`0.3456 -> 0.3560`) because real users consume multiple genres in the held-out test set.
   - Excessively large diversity weights ($\lambda = 0.80$) push the agent toward niche genres at the expense of core preference alignment.

### Limitations
1. **Simulated Online Environment**: Because MovieLens 100K is a static offline dataset, online RL interaction rewards are simulated from a training-derived preference model (combining training SVD collaborative factors, genre affinities, and Bayesian item quality). While held-out test data is strictly isolated, any simulator introduces approximation bias relative to live A/B testing.
2. **Cold-Start Users**: Users with no prior interactions require initial onboarding (addressed in our Streamlit application via interactive genre selection and live like/dislike feedback).
3. **Candidate Pool Scaling**: For catalogs with millions of items, computing Q-values over all items requires two-stage candidate retrieval followed by RL re-ranking.

### Conclusion
**MovieMind** demonstrates that formulating movie recommendation as a diversity-aware Markov Decision Process solved via Deep Q-Networks effectively bridges the gap between recommendation accuracy and catalog exploration, preventing filter bubbles while improving multi-genre user satisfaction.
